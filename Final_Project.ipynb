```python
from IPython.display import Markdown, display

display(Markdown(r"""
# 📘 PART A — CONCEPTUAL FOUNDATION

> **Objective:** Understand the fundamental concepts of ensemble learning before implementing classification and regression models.

---

## 1. What are Ensemble Learning Methods and Why Are They Effective?

**Ensemble learning** is a machine learning approach that combines the predictions of multiple individual models, called base learners, to produce a final prediction.  
Instead of depending on one model, an ensemble uses the strengths of several learners.  
Combining diverse models can reduce individual model weaknesses and improve predictive reliability.  
Ensemble methods can improve **accuracy, stability, robustness, and generalization** on unseen data.  
They are especially useful when different models capture different patterns in the same dataset.

---

## 2. Difference Between Bagging, Boosting, and Voting

| Technique | Core Idea | Learning Strategy | Main Benefit |
|:---|:---|:---|:---|
| **Bagging** | Trains multiple models on different bootstrap samples and combines their predictions. | Parallel | Reduces variance and improves stability |
| **Boosting** | Trains models sequentially, where later models focus on previous errors. | Sequential | Improves weak learners and predictive performance |
| **Voting** | Combines predictions from different base models using majority voting or probabilities. | Parallel Combination | Uses model diversity to improve final predictions |

**Bagging** focuses mainly on reducing variance, while **Boosting** builds models sequentially to correct errors.  
**Voting** combines different models using a predefined voting strategy rather than sequential error correction.

---

## 3. What Is the Bias-Variance Trade-off and How Do Ensemble Methods Address It?

The **bias-variance trade-off** describes the balance between underfitting and overfitting in a machine learning model.  
A model with **high bias** is usually too simple and may underfit the training data.  
A model with **high variance** is overly sensitive to training data and may overfit.  
**Bagging** reduces variance by averaging predictions from multiple learners trained on different samples.  
**Boosting** can reduce bias by sequentially improving weak learners and correcting their errors.  
Therefore, ensemble methods can achieve better generalization on unseen data.

---

## 4. Explain Voting Classifier and Stacking Ensemble

A **Voting Classifier** combines predictions from multiple classification models to make one final prediction.  
**Hard Voting** selects the class receiving the majority of votes, while **Soft Voting** averages the predicted class probabilities.  
**Stacking** also combines multiple base models, but it uses another model called a **meta-learner** to learn how their predictions should be combined.  
The base learners first generate predictions, which are then provided to the meta-learner.  
Therefore, Voting uses a predefined combination rule, whereas Stacking learns the best combination from the data.

---

## 5. Compare AdaBoost, Gradient Boosting, LightGBM, and XGBoost

| Algorithm | Main Concept | Key Strength |
|:---|:---|:---|
| **AdaBoost** | Increases the importance of incorrectly predicted samples for subsequent learners. | Simple and effective sequential boosting |
| **Gradient Boosting** | Sequentially minimizes a loss function by learning from previous prediction errors. | Strong predictive performance |
| **LightGBM** | Uses histogram-based learning and efficient leaf-wise tree growth. | Fast training and memory efficiency |
| **XGBoost** | Uses optimized gradient boosting with regularization and efficient tree construction. | High performance and robustness |

**AdaBoost** focuses on difficult observations, while **Gradient Boosting** optimizes a loss function using gradients.  
**LightGBM** emphasizes computational efficiency and fast tree construction.  
**XGBoost** adds strong regularization and optimization techniques to gradient boosting.  
All four are powerful ensemble techniques, but their implementation strategies and computational characteristics differ.

---

## 🎯 Part A — Key Takeaway

Ensemble learning improves machine learning performance by combining multiple learners instead of depending on a single model.  
**Bagging** mainly improves stability by reducing variance, while **Boosting** improves models sequentially by correcting errors.  
**Voting** combines predictions using predefined voting rules, whereas **Stacking** learns how to combine models using a meta-learner.  
In this project, these ensemble approaches will be implemented and evaluated for both **classification** and **regression** tasks.  
The final objective is to identify the most effective model based on predictive performance, reliability, and practical considerations.
"""))
```

```python
# ============================================================
# PART B — DATASET UNDERSTANDING & PREPARATION
# ============================================================


# ============================================================
# B0. PART B INTRODUCTION
# ============================================================

display(Markdown("""
# 📊 Part B — Dataset Understanding & Preparation

## Objective

Prepare the Smart Outcome Predictor dataset for two supervised
machine-learning tasks:

- **Classification:** Predict `completion_status`
- **Regression:** Predict `final_score`

The preparation workflow includes:

1. Dataset loading
2. Dataset structure inspection
3. Data-quality analysis
4. Duplicate detection
5. Missing-value analysis
6. Date feature engineering
7. Target validation
8. Identifier removal
9. Leakage prevention
10. Train-test splitting
11. Numerical and categorical preprocessing
12. Exploratory data analysis
13. Data-quality validation

> **Data Leakage Prevention:** All preprocessing transformations are
> fitted only on the training dataset and then applied to the test set.
"""))


# ============================================================
# B1. LOAD DATASET
# ============================================================

print("=" * 80)
print("B1. LOADING DATASET")
print("=" * 80)

try:

    df = pd.read_csv(DATA_URL)

    if df.empty:
        raise ValueError("The dataset is empty.")

    print("✅ Dataset loaded successfully!")
    print(
        f"📐 Dataset shape: "
        f"{df.shape[0]:,} rows × {df.shape[1]:,} columns"
    )

except Exception as error:

    print("❌ Dataset loading failed.")
    print("\nPossible causes:")
    print("• Google Sheet is not publicly accessible.")
    print("• Incorrect Sheet ID.")
    print("• Internet connection problem.")
    print("\nOriginal error:")
    print(error)

    raise


# ============================================================
# B2. STANDARDIZE COLUMN NAMES
# ============================================================

print("\n" + "=" * 80)
print("B2. STANDARDIZING COLUMN NAMES")
print("=" * 80)

original_columns = df.columns.tolist()

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r"[^\w\s]", "", regex=True)
    .str.replace(r"\s+", "_", regex=True)
)

print("Original columns:")
print(original_columns)

print("\nStandardized columns:")
print(df.columns.tolist())


# ============================================================
# B3. VALIDATE REQUIRED TARGETS
# ============================================================

print("\n" + "=" * 80)
print("B3. TARGET VALIDATION")
print("=" * 80)

required_targets = [
    CLASSIFICATION_TARGET,
    REGRESSION_TARGET
]

missing_targets = [
    column
    for column in required_targets
    if column not in df.columns
]

if missing_targets:

    raise ValueError(
        f"Required target columns are missing: {missing_targets}\n"
        f"Available columns: {df.columns.tolist()}"
    )

print("✅ Required target variables found.")

print(
    f"   Classification → {CLASSIFICATION_TARGET}"
)

print(
    f"   Regression     → {REGRESSION_TARGET}"
)


# ============================================================
# B4. DATASET PREVIEW
# ============================================================

print("\n" + "=" * 80)
print("B4. DATASET PREVIEW")
print("=" * 80)

display(df.head(10))


# ============================================================
# B5. DATASET INFORMATION
# ============================================================

print("\n" + "=" * 80)
print("B5. DATASET INFORMATION")
print("=" * 80)

df.info()


# ============================================================
# B6. DATA TYPE & QUALITY SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("B6. DATA TYPE & QUALITY SUMMARY")
print("=" * 80)

data_summary = pd.DataFrame({

    "Column": df.columns,

    "Data Type": [
        str(df[column].dtype)
        for column in df.columns
    ],

    "Unique Values": [
        df[column].nunique(dropna=True)
        for column in df.columns
    ],

    "Missing Values": [
        df[column].isna().sum()
        for column in df.columns
    ],

    "Missing %": [
        round(
            df[column].isna().mean() * 100,
            2
        )
        for column in df.columns
    ]
})

display(data_summary)


# ============================================================
# B7. DUPLICATE ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("B7. DUPLICATE ANALYSIS")
print("=" * 80)

duplicate_count = df.duplicated().sum()

print(
    f"🔍 Duplicate rows detected: {duplicate_count:,}"
)

if duplicate_count > 0:

    df = (
        df
        .drop_duplicates()
        .reset_index(drop=True)
    )

    print(
        f"✅ Removed {duplicate_count:,} duplicate rows."
    )

else:

    print("✅ No duplicate rows found.")

print(
    f"📐 Current dataset shape: {df.shape}"
)


# ============================================================
# B8. MISSING VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("B8. MISSING VALUE ANALYSIS")
print("=" * 80)

missing_summary = (
    df.isna()
      .sum()
      .sort_values(ascending=False)
      .to_frame("Missing Values")
)

missing_summary["Missing %"] = (
    missing_summary["Missing Values"]
    / len(df)
    * 100
).round(2)

display(missing_summary)

total_missing = df.isna().sum().sum()

print(
    f"\nTotal missing values: {total_missing:,}"
)

if total_missing > 0:

    print(
        "ℹ️ Missing values will be handled inside the "
        "training preprocessing pipeline."
    )

else:

    print("✅ No missing values detected.")


# ============================================================
# B9. DATE FEATURE ENGINEERING
# ============================================================

print("\n" + "=" * 80)
print("B9. DATE FEATURE ENGINEERING")
print("=" * 80)

DATE_COLUMN = "course_start_date"

if DATE_COLUMN in df.columns:

    print(
        f"📅 Processing date column: {DATE_COLUMN}"
    )

    df[DATE_COLUMN] = pd.to_datetime(
        df[DATE_COLUMN],
        errors="coerce"
    )

    invalid_dates = df[DATE_COLUMN].isna().sum()

    print(
        f"Invalid/missing dates after conversion: "
        f"{invalid_dates:,}"
    )

    # --------------------------------------------------------
    # Extract useful temporal features
    # --------------------------------------------------------

    df["course_start_month"] = (
        df[DATE_COLUMN].dt.month
    )

    df["course_start_day_of_week"] = (
        df[DATE_COLUMN].dt.dayofweek
    )

    df["course_start_quarter"] = (
        df[DATE_COLUMN].dt.quarter
    )

    df["course_start_is_weekend"] = (
        df[DATE_COLUMN]
        .dt.dayofweek
        .isin([5, 6])
        .astype(int)
    )

    # --------------------------------------------------------
    # Raw date is removed after feature extraction
    # --------------------------------------------------------

    df = df.drop(
        columns=[DATE_COLUMN]
    )

    print("\n✅ Date feature engineering completed.")

    print(
        "Created temporal features:"
    )

    print("   • course_start_month")
    print("   • course_start_day_of_week")
    print("   • course_start_quarter")
    print("   • course_start_is_weekend")

    print(
        "\n🚫 Raw course_start_date removed."
    )

else:

    print(
        "ℹ️ course_start_date not found; "
        "date engineering skipped."
    )


# ============================================================
# B10. TARGET ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("B10. TARGET VARIABLE ANALYSIS")
print("=" * 80)


# ------------------------------------------------------------
# Classification target
# ------------------------------------------------------------

print(
    f"\n🎯 Classification Target: "
    f"{CLASSIFICATION_TARGET}"
)

classification_distribution = (
    df[CLASSIFICATION_TARGET]
    .value_counts(dropna=False)
    .sort_index()
    .to_frame("Count")
)

classification_distribution["Percentage"] = (
    classification_distribution["Count"]
    / len(df)
    * 100
).round(2)

display(classification_distribution)


# ------------------------------------------------------------
# Regression target
# ------------------------------------------------------------

print(
    f"\n🎯 Regression Target: "
    f"{REGRESSION_TARGET}"
)

regression_statistics = (
    df[REGRESSION_TARGET]
    .describe()
    .to_frame("Value")
)

display(regression_statistics)


# ============================================================
# B11. TARGET VALIDATION
# ============================================================

print("\n" + "=" * 80)
print("B11. TARGET QUALITY VALIDATION")
print("=" * 80)

# Classification target
y_classification = df[
    CLASSIFICATION_TARGET
].copy()

# Regression target
y_regression = pd.to_numeric(
    df[REGRESSION_TARGET],
    errors="coerce"
)

# Validate classification target
if y_classification.isna().any():

    raise ValueError(
        "❌ Classification target contains missing values."
    )

if y_classification.nunique() != 2:

    raise ValueError(
        "❌ Classification target must contain exactly "
        "two classes."
    )

# Convert classification target to integer
y_classification = (
    pd.to_numeric(
        y_classification,
        errors="coerce"
    )
)

if y_classification.isna().any():

    raise ValueError(
        "❌ Classification target contains non-numeric values."
    )

y_classification = y_classification.astype(int)

# Validate regression target
if y_regression.isna().any():

    raise ValueError(
        "❌ Regression target contains missing or "
        "non-numeric values."
    )

y_regression = y_regression.astype(float)

print("✅ Classification target is valid.")
print(
    f"   Classes: {sorted(y_classification.unique())}"
)

print("\n✅ Regression target is valid.")
print(
    f"   Minimum: {y_regression.min():.2f}"
)

print(
    f"   Maximum: {y_regression.max():.2f}"
)


# ============================================================
# B12. REMOVE INVALID TARGET ROWS
# ============================================================

print("\n" + "=" * 80)
print("B12. TARGET ROW QUALITY CHECK")
print("=" * 80)

valid_target_rows = (
    y_classification.notna()
    &
    y_regression.notna()
)

invalid_target_count = (
    (~valid_target_rows).sum()
)

if invalid_target_count > 0:

    df = (
        df.loc[valid_target_rows]
        .reset_index(drop=True)
    )

    y_classification = (
        y_classification
        .loc[valid_target_rows]
        .reset_index(drop=True)
    )

    y_regression = (
        y_regression
        .loc[valid_target_rows]
        .reset_index(drop=True)
    )

    print(
        f"⚠️ Removed {invalid_target_count:,} "
        "invalid target rows."
    )

else:

    print(
        "✅ All target rows are valid."
    )


# ============================================================
# B13. IDENTIFIER ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("B13. IDENTIFIER ANALYSIS")
print("=" * 80)

ID_COLUMNS = []

for column in df.columns:

    column_lower = column.lower()

    if (
        column_lower == "id"
        or column_lower == "index"
        or column_lower.startswith("unnamed")
        or column_lower.endswith("_id")
    ):

        if column not in {
            CLASSIFICATION_TARGET,
            REGRESSION_TARGET
        }:

            ID_COLUMNS.append(column)

print(
    "Identifier/index columns detected:"
)

if ID_COLUMNS:

    for column in ID_COLUMNS:
        print(f"   🚫 {column}")

else:

    print("   None")


# ============================================================
# B14. BUILD FEATURE MATRIX
# ============================================================

print("\n" + "=" * 80)
print("B14. FEATURE MATRIX & LEAKAGE PREVENTION")
print("=" * 80)

DROP_COLUMNS = list(
    dict.fromkeys(
        [
            CLASSIFICATION_TARGET,
            REGRESSION_TARGET,
            *ID_COLUMNS
        ]
    )
)

X = df.drop(
    columns=DROP_COLUMNS,
    errors="ignore"
).copy()

print(
    "Columns excluded from model features:"
)

for column in DROP_COLUMNS:

    print(
        f"   🚫 {column}"
    )

print(
    f"\n✅ Original predictor count: "
    f"{df.shape[1] - 2}"
)

print(
    f"✅ Final feature count: "
    f"{X.shape[1]}"
)

# ------------------------------------------------------------
# Explicit leakage assertion
# ------------------------------------------------------------

leakage_columns = set(
    X.columns
).intersection(
    {
        CLASSIFICATION_TARGET,
        REGRESSION_TARGET
    }
)

if leakage_columns:

    raise ValueError(
        f"❌ Target leakage detected: "
        f"{leakage_columns}"
    )

print(
    "\n🔐 Target leakage check: PASSED"
)


# ============================================================
# B15. EMPTY FEATURE CHECK
# ============================================================

print("\n" + "=" * 80)
print("B15. EMPTY FEATURE CHECK")
print("=" * 80)

empty_columns = [
    column
    for column in X.columns
    if X[column].isna().all()
]

if empty_columns:

    print(
        "⚠️ Completely empty columns:"
    )

    for column in empty_columns:
        print(f"   🚫 {column}")

    X = X.drop(
        columns=empty_columns
    )

else:

    print(
        "✅ No completely empty predictor columns."
    )


# ============================================================
# B16. CLASSIFICATION TRAIN-TEST SPLIT
# ============================================================

print("\n" + "=" * 80)
print("B16. CLASSIFICATION TRAIN-TEST SPLIT")
print("=" * 80)

(
    X_train_cls,
    X_test_cls,
    y_train_cls,
    y_test_cls
) = train_test_split(
    X,
    y_classification,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y_classification
)

print(
    f"Training samples : {len(X_train_cls):,}"
)

print(
    f"Testing samples  : {len(X_test_cls):,}"
)

classification_split = pd.DataFrame({

    "Train %": (
        y_train_cls
        .value_counts(normalize=True)
        .sort_index()
        * 100
    ).round(2),

    "Test %": (
        y_test_cls
        .value_counts(normalize=True)
        .sort_index()
        * 100
    ).round(2)
})

display(classification_split)


# ============================================================
# B17. REGRESSION TRAIN-TEST SPLIT
# ============================================================

print("\n" + "=" * 80)
print("B17. REGRESSION TRAIN-TEST SPLIT")
print("=" * 80)

(
    X_train_reg,
    X_test_reg,
    y_train_reg,
    y_test_reg
) = train_test_split(
    X,
    y_regression,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE
)

print(
    f"Training samples : {len(X_train_reg):,}"
)

print(
    f"Testing samples  : {len(X_test_reg):,}"
)


# ============================================================
# B18. FEATURE TYPE DETECTION
# ============================================================

print("\n" + "=" * 80)
print("B18. FEATURE TYPE DETECTION")
print("=" * 80)

numeric_features = (
    X_train_cls
    .select_dtypes(include=np.number)
    .columns
    .tolist()
)

categorical_features = (
    X_train_cls
    .select_dtypes(exclude=np.number)
    .columns
    .tolist()
)

print(
    f"🔢 Numeric features     : "
    f"{len(numeric_features)}"
)

print(
    f"🔤 Categorical features : "
    f"{len(categorical_features)}"
)

print("\nNumeric features:")

for feature in numeric_features:
    print(f"   • {feature}")

print("\nCategorical features:")

for feature in categorical_features:
    print(f"   • {feature}")


# ============================================================
# B19. VERSION-COMPATIBLE ONE-HOT ENCODER
# ============================================================

print("\n" + "=" * 80)
print("B19. PREPROCESSING PIPELINE")
print("=" * 80)

try:

    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )

except TypeError:

    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )


# ------------------------------------------------------------
# Numerical pipeline
# ------------------------------------------------------------

numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        )
    ]
)


# ------------------------------------------------------------
# Categorical pipeline
# ------------------------------------------------------------

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),

        (
            "encoder",
            encoder
        )
    ]
)


# ------------------------------------------------------------
# Build transformer list dynamically
# ------------------------------------------------------------

transformers = []

if numeric_features:

    transformers.append(
        (
            "numeric",
            numeric_pipeline,
            numeric_features
        )
    )

if categorical_features:

    transformers.append(
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    )


# ------------------------------------------------------------
# Separate preprocessors
# ------------------------------------------------------------

preprocessor_cls = ColumnTransformer(
    transformers=transformers,
    remainder="drop"
)

preprocessor_reg = ColumnTransformer(
    transformers=transformers,
    remainder="drop"
)

print("✅ Numerical preprocessing:")
print("   • Median imputation")
print("   • Standard scaling")

print("\n✅ Categorical preprocessing:")
print("   • Most-frequent imputation")
print("   • One-hot encoding")
print("   • Unknown categories ignored")


# ============================================================
# B20. FIT CLASSIFICATION PREPROCESSOR
# ============================================================

print("\n" + "=" * 80)
print("B20. CLASSIFICATION PREPROCESSING")
print("=" * 80)

X_train_cls_processed = (
    preprocessor_cls
    .fit_transform(X_train_cls)
)

X_test_cls_processed = (
    preprocessor_cls
    .transform(X_test_cls)
)

print(
    "✅ Preprocessor fitted ONLY on classification training data."
)

print(
    f"Processed training shape: "
    f"{X_train_cls_processed.shape}"
)

print(
    f"Processed testing shape: "
    f"{X_test_cls_processed.shape}"
)


# ============================================================
# B21. FIT REGRESSION PREPROCESSOR
# ============================================================

print("\n" + "=" * 80)
print("B21. REGRESSION PREPROCESSING")
print("=" * 80)

X_train_reg_processed = (
    preprocessor_reg
    .fit_transform(X_train_reg)
)

X_test_reg_processed = (
    preprocessor_reg
    .transform(X_test_reg)
)

print(
    "✅ Preprocessor fitted ONLY on regression training data."
)

print(
    f"Processed training shape: "
    f"{X_train_reg_processed.shape}"
)

print(
    f"Processed testing shape: "
    f"{X_test_reg_processed.shape}"
)


# ============================================================
# B22. PROCESSED FEATURE NAMES
# ============================================================

print("\n" + "=" * 80)
print("B22. PROCESSED FEATURE INFORMATION")
print("=" * 80)

processed_feature_names = (
    preprocessor_cls
    .get_feature_names_out()
    .tolist()
)

print(
    f"Original predictor count : "
    f"{X.shape[1]}"
)

print(
    f"Processed feature count  : "
    f"{len(processed_feature_names)}"
)

print("\nFirst 30 processed features:")

for feature in processed_feature_names[:30]:

    print(
        f"   • {feature}"
    )

if len(processed_feature_names) > 30:

    print(
        f"\n   ... and "
        f"{len(processed_feature_names) - 30} more."
    )


# ============================================================
# B23. CLASSIFICATION TARGET VISUALIZATION
# ============================================================

display(Markdown("""
## 📈 Exploratory Data Analysis

The following visualizations examine the target variables and
important numerical predictors before ensemble model training.
"""))


plt.figure(figsize=(8, 5))

sns.countplot(
    x=y_classification
)

plt.title(
    "Completion Status Distribution",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel(
    CLASSIFICATION_TARGET
)

plt.ylabel(
    "Number of Students"
)

plt.tight_layout()
plt.show()


# ============================================================
# B24. REGRESSION TARGET DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    y_regression,
    bins=25,
    kde=True
)

plt.title(
    "Final Score Distribution",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel(
    REGRESSION_TARGET
)

plt.ylabel(
    "Number of Students"
)

plt.tight_layout()
plt.show()


# ============================================================
# B25. MISSING VALUE VISUALIZATION
# ============================================================

missing_plot = (
    df.isna()
      .sum()
      .sort_values(ascending=False)
)

missing_plot = (
    missing_plot[
        missing_plot > 0
    ]
)

if len(missing_plot) > 0:

    plt.figure(figsize=(9, 5))

    sns.barplot(
        x=missing_plot.values,
        y=missing_plot.index
    )

    plt.title(
        "Missing Values by Feature",
        fontsize=14,
        fontweight="bold"
    )

    plt.xlabel(
        "Number of Missing Values"
    )

    plt.ylabel(
        "Feature"
    )

    plt.tight_layout()
    plt.show()

else:

    print(
        "✅ No missing-value plot required."
    )


# ============================================================
# B26. NUMERICAL FEATURE CORRELATION
# ============================================================

numeric_analysis = df.select_dtypes(
    include=np.number
)

numeric_analysis = numeric_analysis.drop(
    columns=[
        CLASSIFICATION_TARGET,
        REGRESSION_TARGET
    ],
    errors="ignore"
)

if numeric_analysis.shape[1] >= 2:

    correlation_matrix = (
        numeric_analysis
        .corr()
    )

    plt.figure(
        figsize=(
            max(
                9,
                numeric_analysis.shape[1] * 0.7
            ),
            max(
                7,
                numeric_analysis.shape[1] * 0.6
            )
        )
    )

    sns.heatmap(
        correlation_matrix,
        annot=(
            numeric_analysis.shape[1] <= 12
        ),
        fmt=".2f",
        cmap="coolwarm",
        center=0
    )

    plt.title(
        "Numerical Feature Correlation Matrix",
        fontsize=14,
        fontweight="bold"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# B27. CLASSIFICATION — KEY NUMERIC FEATURES
# ============================================================

key_numeric_features = [
    feature
    for feature in [
        "attendance_rate",
        "avg_quiz_score",
        "time_spent_hours",
        "sessions"
    ]
    if feature in df.columns
]

for feature in key_numeric_features:

    plt.figure(figsize=(8, 5))

    sns.boxplot(
        x=df[CLASSIFICATION_TARGET],
        y=df[feature]
    )

    plt.title(
        f"{feature.replace('_', ' ').title()} "
        f"vs Completion Status",
        fontsize=14,
        fontweight="bold"
    )

    plt.xlabel(
        CLASSIFICATION_TARGET
    )

    plt.ylabel(
        feature.replace("_", " ").title()
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# B28. REGRESSION — KEY NUMERIC FEATURES
# ============================================================

for feature in key_numeric_features:

    plt.figure(figsize=(8, 5))

    sns.scatterplot(
        data=df,
        x=feature,
        y=REGRESSION_TARGET,
        alpha=0.5
    )

    plt.title(
        f"{feature.replace('_', ' ').title()} "
        f"vs Final Score",
        fontsize=14,
        fontweight="bold"
    )

    plt.xlabel(
        feature.replace("_", " ").title()
    )

    plt.ylabel(
        REGRESSION_TARGET.replace("_", " ").title()
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# B29. FINAL DATA QUALITY CHECK
# ============================================================

print("\n" + "=" * 80)
print("B29. FINAL DATA QUALITY GATE")
print("=" * 80)

# Dataset checks
assert len(X) == len(y_classification)
assert len(X) == len(y_regression)

# Split checks
assert len(X_train_cls) == len(y_train_cls)
assert len(X_test_cls) == len(y_test_cls)

assert len(X_train_reg) == len(y_train_reg)
assert len(X_test_reg) == len(y_test_reg)

# Leakage checks
assert CLASSIFICATION_TARGET not in X.columns
assert REGRESSION_TARGET not in X.columns

for column in ID_COLUMNS:

    assert column not in X.columns


# Target checks
assert y_train_cls.notna().all()
assert y_test_cls.notna().all()

assert y_train_reg.notna().all()
assert y_test_reg.notna().all()

# Processed-data checks
assert X_train_cls_processed.shape[0] == len(y_train_cls)
assert X_test_cls_processed.shape[0] == len(y_test_cls)

assert X_train_reg_processed.shape[0] == len(y_train_reg)
assert X_test_reg_processed.shape[0] == len(y_test_reg)


print("✅ Dataset integrity check       : PASSED")
print("✅ Target validation             : PASSED")
print("✅ Duplicate validation          : PASSED")
print("✅ Leakage prevention            : PASSED")
print("✅ Classification split          : PASSED")
print("✅ Regression split              : PASSED")
print("✅ Missing-value strategy        : PASSED")
print("✅ Feature preprocessing         : PASSED")
print("✅ Processed-data dimensions     : PASSED")


# ============================================================
# B30. EXPORT CLEANED DATASET
# ============================================================

print("\n" + "=" * 80)
print("B30. EXPORTING CLEANED DATASET")
print("=" * 80)

cleaned_dataset_path = (
    "smart_outcome_cleaned.csv"
)

df.to_csv(
    cleaned_dataset_path,
    index=False
)

print(
    f"✅ Cleaned dataset exported to: "
    f"{cleaned_dataset_path}"
)


# ============================================================
# B31. FINAL PART B REPORT
# ============================================================

display(Markdown(f"""
---

# ✅ Part B — Completed Successfully

| Component | Status |
|---|---|
| Dataset loading | ✅ |
| Dataset inspection | ✅ |
| Column standardization | ✅ |
| Duplicate detection | ✅ |
| Missing-value analysis | ✅ |
| Date feature engineering | ✅ |
| Target validation | ✅ |
| Identifier removal | ✅ |
| Leakage prevention | ✅ |
| Classification split | ✅ |
| Regression split | ✅ |
| Numerical preprocessing | ✅ |
| Categorical preprocessing | ✅ |
| One-hot encoding | ✅ |
| Feature scaling | ✅ |
| Target EDA | ✅ |
| Predictor EDA | ✅ |
| Data-quality validation | ✅ |
| Cleaned dataset export | ✅ |

## 🎯 Prediction Tasks

**Classification target:** `{CLASSIFICATION_TARGET}`

**Regression target:** `{REGRESSION_TARGET}`

## 📊 Dataset

**Rows:** {len(df):,}

**Original columns:** 19

**Model predictor columns:** {X.shape[1]}

**Classification training samples:** {len(X_train_cls):,}

**Classification testing samples:** {len(X_test_cls):,}

**Regression training samples:** {len(X_train_reg):,}

**Regression testing samples:** {len(X_test_reg):,}

**Processed classification features:** {X_train_cls_processed.shape[1]:,}

**Processed regression features:** {X_train_reg_processed.shape[1]:,}

## 🔐 Leakage Prevention

The two target variables and identifier columns were excluded from
the feature matrix. Preprocessing was fitted only on training data
and subsequently applied to unseen test data.

## 🚀 Ready for Part C

The prepared datasets are now ready for:

- Bagging Classifier
- Bagging Regressor
- Single-model baseline comparison
- Cross-validation
- Ensemble performance evaluation
"""))
```

```python
# =============================================================================
# 🧩 PART C — BAGGING ENSEMBLE METHODS & BASELINE COMPARISON
# =============================================================================
#
# SMART OUTCOME PREDICTOR
#
# Part C Requirements:
#   1. Single Decision Tree Classifier
#   2. Bagging Classifier
#   3. Compare classification performance
#   4. 5-Fold Cross-Validation
#   5. Confusion Matrix
#   6. ROC Curve
#   7. Single Decision Tree Regressor
#   8. Bagging Regressor
#   9. Compare regression performance
#  10. 5-Fold Cross-Validation
#  11. Actual vs Predicted
#  12. Residual Analysis
#  13. Feature Importance
#  14. Training Efficiency
#  15. Automatic conclusion
#
# NOTE:
# All major libraries are already imported in Part A.
# No ConfusionMatrixDisplay is required.
# =============================================================================


# =============================================================================
# C1 — INTRODUCTION
# =============================================================================

display(Markdown("""
# 🧩 Part C — Bagging Ensemble Methods & Baseline Comparison

## Objective

Bagging, or **Bootstrap Aggregating**, is an ensemble-learning technique
that combines multiple models trained on different bootstrap samples of
the training data.

This section compares:

### Classification
- Decision Tree Classifier
- Bagging Classifier

### Regression
- Decision Tree Regressor
- Bagging Regressor

### Evaluation
- Accuracy
- Precision
- Recall
- F1-Score
- ROC-AUC
- MAE
- RMSE
- R²
- 5-Fold Cross-Validation
- Confusion Matrix
- ROC Curve
- Actual vs Predicted Plot
- Residual Analysis
- Feature Importance
- Training Time

The objective is to determine whether Bagging improves model stability
and generalization compared with a single Decision Tree.
"""))


# =============================================================================
# C2 — VALIDATE PART B OUTPUT
# =============================================================================

print("=" * 85)
print("C2 — VALIDATING PART B OUTPUT")
print("=" * 85)

required_objects = [
    "X_train_cls",
    "X_test_cls",
    "y_train_cls",
    "y_test_cls",
    "X_train_reg",
    "X_test_reg",
    "y_train_reg",
    "y_test_reg"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        "The following Part B objects are missing:\n\n"
        + "\n".join(f"• {obj}" for obj in missing_objects)
        + "\n\nPlease execute Part B completely before Part C."
    )

print("✅ All required Part B objects are available.")


# =============================================================================
# C3 — IDENTIFY PREPROCESSORS
# =============================================================================

print("\n" + "=" * 85)
print("C3 — IDENTIFYING PREPROCESSING PIPELINES")
print("=" * 85)

# -------------------------------------------------------------------------
# Part B may have created these preprocessors under different names.
# We detect the common names automatically.
# -------------------------------------------------------------------------

classification_preprocessor = None
regression_preprocessor = None

for name in [
    "preprocessor_cls",
    "classification_preprocessor",
    "cls_preprocessor"
]:
    if name in globals():
        classification_preprocessor = globals()[name]
        break


for name in [
    "preprocessor_reg",
    "regression_preprocessor",
    "reg_preprocessor"
]:
    if name in globals():
        regression_preprocessor = globals()[name]
        break


if classification_preprocessor is not None:
    print("✅ Classification preprocessor detected.")
else:
    print("ℹ️ Classification preprocessor object not separately detected.")


if regression_preprocessor is not None:
    print("✅ Regression preprocessor detected.")
else:
    print("ℹ️ Regression preprocessor object not separately detected.")


# =============================================================================
# C4 — CONVERT DATA INTO NUMERIC MODEL MATRICES
# =============================================================================

print("\n" + "=" * 85)
print("C4 — MODEL INPUT VALIDATION & NUMERIC CONVERSION")
print("=" * 85)


def make_numeric_matrix(
    X_train,
    X_test,
    preprocessor,
    task_name
):
    """
    Ensures that Decision Tree and Bagging models receive numeric data.

    If X_train/X_test are already numeric, they are used directly.

    If categorical/object columns remain, the fitted Part B preprocessor
    is used to transform them.
    """

    # ---------------------------------------------------------------------
    # Case 1 — pandas DataFrame containing categorical/object columns
    # ---------------------------------------------------------------------

    if isinstance(X_train, pd.DataFrame):

        object_columns = X_train.select_dtypes(
            include=["object", "category"]
        ).columns.tolist()

        if len(object_columns) > 0:

            print(
                f"⚠️ {task_name}: "
                f"{len(object_columns)} categorical columns detected."
            )

            if preprocessor is None:

                raise ValueError(
                    f"{task_name}: categorical features are present, "
                    "but no fitted Part B preprocessor was found."
                )

            X_train_processed = preprocessor.transform(
                X_train
            )

            X_test_processed = preprocessor.transform(
                X_test
            )

            print(
                f"✅ {task_name}: Part B preprocessing applied."
            )

        else:

            X_train_processed = X_train
            X_test_processed = X_test

            print(
                f"✅ {task_name}: data is already numeric."
            )

    # ---------------------------------------------------------------------
    # Case 2 — already transformed NumPy / sparse matrix
    # ---------------------------------------------------------------------

    else:

        X_train_processed = X_train
        X_test_processed = X_test

        print(
            f"✅ {task_name}: processed matrix detected."
        )

    # ---------------------------------------------------------------------
    # Convert sparse matrices only for validation.
    # Keep sparse representation for model training when possible.
    # ---------------------------------------------------------------------

    if hasattr(X_train_processed, "toarray"):

        train_validation = X_train_processed.toarray()
        test_validation = X_test_processed.toarray()

    else:

        train_validation = np.asarray(
            X_train_processed
        )

        test_validation = np.asarray(
            X_test_processed
        )

    # ---------------------------------------------------------------------
    # Numeric validation
    # ---------------------------------------------------------------------

    if not np.isfinite(train_validation).all():

        raise ValueError(
            f"{task_name}: training data contains NaN or infinite values."
        )

    if not np.isfinite(test_validation).all():

        raise ValueError(
            f"{task_name}: testing data contains NaN or infinite values."
        )

    return (
        X_train_processed,
        X_test_processed
    )


# -------------------------------------------------------------------------
# Classification
# -------------------------------------------------------------------------

X_train_cls_model, X_test_cls_model = make_numeric_matrix(
    X_train_cls,
    X_test_cls,
    classification_preprocessor,
    "Classification"
)


# -------------------------------------------------------------------------
# Regression
# -------------------------------------------------------------------------

X_train_reg_model, X_test_reg_model = make_numeric_matrix(
    X_train_reg,
    X_test_reg,
    regression_preprocessor,
    "Regression"
)


# =============================================================================
# C5 — FINAL DATA QUALITY CHECK
# =============================================================================

print("\n" + "=" * 85)
print("C5 — FINAL MODEL INPUT QUALITY CHECK")
print("=" * 85)

assert X_train_cls_model.shape[0] == len(y_train_cls)
assert X_test_cls_model.shape[0] == len(y_test_cls)

assert X_train_reg_model.shape[0] == len(y_train_reg)
assert X_test_reg_model.shape[0] == len(y_test_reg)

assert X_train_cls_model.shape[1] == X_test_cls_model.shape[1]
assert X_train_reg_model.shape[1] == X_test_reg_model.shape[1]

print(
    f"Classification Train : {X_train_cls_model.shape}"
)

print(
    f"Classification Test  : {X_test_cls_model.shape}"
)

print(
    f"Regression Train     : {X_train_reg_model.shape}"
)

print(
    f"Regression Test      : {X_test_reg_model.shape}"
)

print("\n✅ Row-count validation      : PASSED")
print("✅ Feature-count validation  : PASSED")
print("✅ Numeric validation        : PASSED")
print("✅ Model input quality gate  : PASSED")


# =============================================================================
# C6 — BAGGING THEORY
# =============================================================================

display(Markdown("""
## C6 — Bagging: Bootstrap Aggregating

Bagging trains multiple base learners using different bootstrap samples
of the training dataset.

Each learner is trained independently.

### Classification

Predictions are combined using **majority voting**.

### Regression

Predictions are combined using **averaging**.

### Why Bagging?

Decision Trees can have high variance. Bagging reduces this variance by
combining several independently trained trees.

Therefore, Bagging can provide:

- Reduced variance
- More stable predictions
- Better generalization
- Greater robustness to training-sample variation
"""))


# =============================================================================
# C7 — MODEL CONFIGURATION
# =============================================================================

print("=" * 85)
print("C7 — BAGGING MODEL CONFIGURATION")
print("=" * 85)

C_RANDOM_STATE = RANDOM_STATE
C_CV_FOLDS = CV_FOLDS
C_N_ESTIMATORS = 100

print(f"Random State           : {C_RANDOM_STATE}")
print(f"Cross-Validation       : {C_CV_FOLDS}-Fold")
print(f"Number of Estimators   : {C_N_ESTIMATORS}")
print("Bootstrap Sampling     : Enabled")
print("Parallel Processing    : Enabled")


# =============================================================================
# C8 — DECISION TREE CLASSIFIER BASELINE
# =============================================================================

print("\n" + "=" * 85)
print("C8 — SINGLE DECISION TREE CLASSIFIER")
print("=" * 85)

if "time" not in globals():
    import time


dt_classifier = DecisionTreeClassifier(
    random_state=C_RANDOM_STATE
)


start_time = time.perf_counter()

dt_classifier.fit(
    X_train_cls_model,
    y_train_cls
)

dt_cls_train_time = (
    time.perf_counter() - start_time
)


dt_cls_pred = dt_classifier.predict(
    X_test_cls_model
)

dt_cls_prob = dt_classifier.predict_proba(
    X_test_cls_model
)[:, 1]


dt_cls_accuracy = accuracy_score(
    y_test_cls,
    dt_cls_pred
)

dt_cls_precision = precision_score(
    y_test_cls,
    dt_cls_pred,
    zero_division=0
)

dt_cls_recall = recall_score(
    y_test_cls,
    dt_cls_pred,
    zero_division=0
)

dt_cls_f1 = f1_score(
    y_test_cls,
    dt_cls_pred,
    zero_division=0
)

dt_cls_roc_auc = roc_auc_score(
    y_test_cls,
    dt_cls_prob
)


print("🌳 Decision Tree Classifier trained.")
print()
print(f"Training Time : {dt_cls_train_time:.4f} sec")
print(f"Accuracy      : {dt_cls_accuracy:.4f}")
print(f"Precision     : {dt_cls_precision:.4f}")
print(f"Recall        : {dt_cls_recall:.4f}")
print(f"F1-Score      : {dt_cls_f1:.4f}")
print(f"ROC-AUC       : {dt_cls_roc_auc:.4f}")


# =============================================================================
# C9 — BAGGING CLASSIFIER
# =============================================================================

print("\n" + "=" * 85)
print("C9 — BAGGING CLASSIFIER")
print("=" * 85)


bag_cls_base = DecisionTreeClassifier(
    random_state=C_RANDOM_STATE
)


bag_classifier = BaggingClassifier(
    estimator=bag_cls_base,
    n_estimators=C_N_ESTIMATORS,
    max_samples=1.0,
    max_features=1.0,
    bootstrap=True,
    bootstrap_features=False,
    oob_score=True,
    n_jobs=-1,
    random_state=C_RANDOM_STATE
)


start_time = time.perf_counter()

bag_classifier.fit(
    X_train_cls_model,
    y_train_cls
)

bag_cls_train_time = (
    time.perf_counter() - start_time
)


bag_cls_pred = bag_classifier.predict(
    X_test_cls_model
)

bag_cls_prob = bag_classifier.predict_proba(
    X_test_cls_model
)[:, 1]


bag_cls_accuracy = accuracy_score(
    y_test_cls,
    bag_cls_pred
)

bag_cls_precision = precision_score(
    y_test_cls,
    bag_cls_pred,
    zero_division=0
)

bag_cls_recall = recall_score(
    y_test_cls,
    bag_cls_pred,
    zero_division=0
)

bag_cls_f1 = f1_score(
    y_test_cls,
    bag_cls_pred,
    zero_division=0
)

bag_cls_roc_auc = roc_auc_score(
    y_test_cls,
    bag_cls_prob
)


print("🌲 Bagging Classifier trained.")
print()
print(f"Estimators    : {C_N_ESTIMATORS}")
print(f"Training Time : {bag_cls_train_time:.4f} sec")
print(f"OOB Score     : {bag_classifier.oob_score_:.4f}")
print(f"Accuracy      : {bag_cls_accuracy:.4f}")
print(f"Precision     : {bag_cls_precision:.4f}")
print(f"Recall        : {bag_cls_recall:.4f}")
print(f"F1-Score      : {bag_cls_f1:.4f}")
print(f"ROC-AUC       : {bag_cls_roc_auc:.4f}")


# =============================================================================
# C10 — CLASSIFICATION COMPARISON
# =============================================================================

print("\n" + "=" * 85)
print("C10 — CLASSIFICATION PERFORMANCE COMPARISON")
print("=" * 85)


classification_comparison = pd.DataFrame({

    "Model": [
        "Decision Tree",
        "Bagging Classifier"
    ],

    "Accuracy": [
        dt_cls_accuracy,
        bag_cls_accuracy
    ],

    "Precision": [
        dt_cls_precision,
        bag_cls_precision
    ],

    "Recall": [
        dt_cls_recall,
        bag_cls_recall
    ],

    "F1-Score": [
        dt_cls_f1,
        bag_cls_f1
    ],

    "ROC-AUC": [
        dt_cls_roc_auc,
        bag_cls_roc_auc
    ],

    "Training Time (sec)": [
        dt_cls_train_time,
        bag_cls_train_time
    ]

}).set_index("Model")


display(classification_comparison.round(4))


# =============================================================================
# C11 — CLASSIFICATION CROSS VALIDATION
# =============================================================================

print("\n" + "=" * 85)
print("C11 — 5-FOLD CLASSIFICATION CROSS-VALIDATION")
print("=" * 85)


classification_cv_models = {

    "Decision Tree": DecisionTreeClassifier(
        random_state=C_RANDOM_STATE
    ),

    "Bagging Classifier": BaggingClassifier(
        estimator=DecisionTreeClassifier(
            random_state=C_RANDOM_STATE
        ),
        n_estimators=C_N_ESTIMATORS,
        bootstrap=True,
        oob_score=True,
        n_jobs=-1,
        random_state=C_RANDOM_STATE
    )
}


classification_cv_results = []


for model_name, model in classification_cv_models.items():

    cv_result = cross_validate(

        model,

        X_train_cls_model,

        y_train_cls,

        cv=C_CV_FOLDS,

        scoring={
            "Accuracy": "accuracy",
            "Precision": "precision",
            "Recall": "recall",
            "F1": "f1",
            "ROC_AUC": "roc_auc"
        },

        n_jobs=-1,

        return_train_score=False
    )


    classification_cv_results.append({

        "Model": model_name,

        "CV Accuracy":
            cv_result["test_Accuracy"].mean(),

        "CV Precision":
            cv_result["test_Precision"].mean(),

        "CV Recall":
            cv_result["test_Recall"].mean(),

        "CV F1-Score":
            cv_result["test_F1"].mean(),

        "CV ROC-AUC":
            cv_result["test_ROC_AUC"].mean(),

        "Accuracy Std":
            cv_result["test_Accuracy"].std()
    })


classification_cv_df = pd.DataFrame(
    classification_cv_results
).set_index("Model")


display(classification_cv_df.round(4))

print("✅ 5-Fold classification CV completed.")


# =============================================================================
# C12 — CONFUSION MATRIX
# =============================================================================

print("\n" + "=" * 85)
print("C12 — CONFUSION MATRIX COMPARISON")
print("=" * 85)


dt_cm = confusion_matrix(
    y_test_cls,
    dt_cls_pred
)


bag_cm = confusion_matrix(
    y_test_cls,
    bag_cls_pred
)


fig, axes = plt.subplots(
    1,
    2,
    figsize=(13, 5)
)


sns.heatmap(
    dt_cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    cbar=False,
    ax=axes[0]
)

axes[0].set_title(
    "Decision Tree Classifier"
)

axes[0].set_xlabel(
    "Predicted Label"
)

axes[0].set_ylabel(
    "Actual Label"
)


sns.heatmap(
    bag_cm,
    annot=True,
    fmt="d",
    cmap="Greens",
    cbar=False,
    ax=axes[1]
)

axes[1].set_title(
    "Bagging Classifier"
)

axes[1].set_xlabel(
    "Predicted Label"
)

axes[1].set_ylabel(
    "Actual Label"
)


plt.tight_layout()
plt.show()


# =============================================================================
# C13 — ROC CURVE
# =============================================================================

print("=" * 85)
print("C13 — ROC CURVE COMPARISON")
print("=" * 85)


dt_fpr, dt_tpr, _ = roc_curve(
    y_test_cls,
    dt_cls_prob
)


bag_fpr, bag_tpr, _ = roc_curve(
    y_test_cls,
    bag_cls_prob
)


plt.figure(figsize=(9, 6))


plt.plot(
    dt_fpr,
    dt_tpr,
    linewidth=2,
    label=(
        f"Decision Tree "
        f"(AUC = {dt_cls_roc_auc:.3f})"
    )
)


plt.plot(
    bag_fpr,
    bag_tpr,
    linewidth=2,
    label=(
        f"Bagging "
        f"(AUC = {bag_cls_roc_auc:.3f})"
    )
)


plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.5,
    label="Random Classifier"
)


plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "ROC Curve — Classification Models"
)

plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()


# =============================================================================
# C14 — DECISION TREE REGRESSOR BASELINE
# =============================================================================

print("\n" + "=" * 85)
print("C14 — SINGLE DECISION TREE REGRESSOR")
print("=" * 85)


dt_regressor = DecisionTreeRegressor(
    random_state=C_RANDOM_STATE
)


start_time = time.perf_counter()


dt_regressor.fit(
    X_train_reg_model,
    y_train_reg
)


dt_reg_train_time = (
    time.perf_counter() - start_time
)


dt_reg_pred = dt_regressor.predict(
    X_test_reg_model
)


dt_reg_mae = mean_absolute_error(
    y_test_reg,
    dt_reg_pred
)


dt_reg_rmse = np.sqrt(
    mean_squared_error(
        y_test_reg,
        dt_reg_pred
    )
)


dt_reg_r2 = r2_score(
    y_test_reg,
    dt_reg_pred
)


print("🌳 Decision Tree Regressor trained.")
print()
print(f"Training Time : {dt_reg_train_time:.4f} sec")
print(f"MAE           : {dt_reg_mae:.4f}")
print(f"RMSE          : {dt_reg_rmse:.4f}")
print(f"R²            : {dt_reg_r2:.4f}")


# =============================================================================
# C15 — BAGGING REGRESSOR
# =============================================================================

print("\n" + "=" * 85)
print("C15 — BAGGING REGRESSOR")
print("=" * 85)


bag_reg_base = DecisionTreeRegressor(
    random_state=C_RANDOM_STATE
)


bag_regressor = BaggingRegressor(
    estimator=bag_reg_base,
    n_estimators=C_N_ESTIMATORS,
    max_samples=1.0,
    max_features=1.0,
    bootstrap=True,
    bootstrap_features=False,
    oob_score=True,
    n_jobs=-1,
    random_state=C_RANDOM_STATE
)


start_time = time.perf_counter()


bag_regressor.fit(
    X_train_reg_model,
    y_train_reg
)


bag_reg_train_time = (
    time.perf_counter() - start_time
)


bag_reg_pred = bag_regressor.predict(
    X_test_reg_model
)


bag_reg_mae = mean_absolute_error(
    y_test_reg,
    bag_reg_pred
)


bag_reg_rmse = np.sqrt(
    mean_squared_error(
        y_test_reg,
        bag_reg_pred
    )
)


bag_reg_r2 = r2_score(
    y_test_reg,
    bag_reg_pred
)


print("🌲 Bagging Regressor trained.")
print()
print(f"Estimators    : {C_N_ESTIMATORS}")
print(f"Training Time : {bag_reg_train_time:.4f} sec")
print(f"OOB Score     : {bag_regressor.oob_score_:.4f}")
print(f"MAE           : {bag_reg_mae:.4f}")
print(f"RMSE          : {bag_reg_rmse:.4f}")
print(f"R²            : {bag_reg_r2:.4f}")


# =============================================================================
# C16 — REGRESSION COMPARISON
# =============================================================================

print("\n" + "=" * 85)
print("C16 — REGRESSION PERFORMANCE COMPARISON")
print("=" * 85)


regression_comparison = pd.DataFrame({

    "Model": [
        "Decision Tree",
        "Bagging Regressor"
    ],

    "MAE": [
        dt_reg_mae,
        bag_reg_mae
    ],

    "RMSE": [
        dt_reg_rmse,
        bag_reg_rmse
    ],

    "R²": [
        dt_reg_r2,
        bag_reg_r2
    ],

    "Training Time (sec)": [
        dt_reg_train_time,
        bag_reg_train_time
    ]

}).set_index("Model")


display(regression_comparison.round(4))

# =============================================================================
# C17 — REGRESSION CROSS VALIDATION
# =============================================================================

print("\n" + "=" * 85)
print("C17 — 5-FOLD REGRESSION CROSS-VALIDATION")
print("=" * 85)


regression_cv_models = {

    "Decision Tree": DecisionTreeRegressor(
        random_state=C_RANDOM_STATE
    ),

    "Bagging Regressor": BaggingRegressor(
        estimator=DecisionTreeRegressor(
            random_state=C_RANDOM_STATE
        ),
        n_estimators=C_N_ESTIMATORS,
        bootstrap=True,
        oob_score=True,
        n_jobs=-1,
        random_state=C_RANDOM_STATE
    )
}


regression_cv_results = []


for model_name, model in regression_cv_models.items():

    cv_result = cross_validate(

        model,

        X_train_reg_model,

        y_train_reg,

        cv=C_CV_FOLDS,

        scoring={
            "MAE": "neg_mean_absolute_error",
            "RMSE": "neg_root_mean_squared_error",
            "R2": "r2"
        },

        n_jobs=-1,

        return_train_score=False
    )


    regression_cv_results.append({

        "Model": model_name,

        "CV MAE":
            -cv_result["test_MAE"].mean(),

        "CV RMSE":
            -cv_result["test_RMSE"].mean(),

        "CV R²":
            cv_result["test_R2"].mean(),

        "R² Std":
            cv_result["test_R2"].std()
    })


regression_cv_df = pd.DataFrame(
    regression_cv_results
).set_index("Model")


display(
    regression_cv_df.style.format(
        "{:.4f}"
    )
)

print("✅ 5-Fold regression CV completed.")


# =============================================================================
# C18 — ACTUAL VS PREDICTED
# =============================================================================

print("\n" + "=" * 85)
print("C18 — ACTUAL VS PREDICTED ANALYSIS")
print("=" * 85)


fig, axes = plt.subplots(
    1,
    2,
    figsize=(13, 5)
)


axes[0].scatter(
    y_test_reg,
    dt_reg_pred,
    alpha=0.55
)

axes[0].plot(
    [y_test_reg.min(), y_test_reg.max()],
    [y_test_reg.min(), y_test_reg.max()],
    linestyle="--",
    linewidth=2
)

axes[0].set_title(
    "Decision Tree Regressor"
)

axes[0].set_xlabel(
    "Actual Final Score"
)

axes[0].set_ylabel(
    "Predicted Final Score"
)

axes[0].grid(alpha=0.25)


axes[1].scatter(
    y_test_reg,
    bag_reg_pred,
    alpha=0.55
)

axes[1].plot(
    [y_test_reg.min(), y_test_reg.max()],
    [y_test_reg.min(), y_test_reg.max()],
    linestyle="--",
    linewidth=2
)

axes[1].set_title(
    "Bagging Regressor"
)

axes[1].set_xlabel(
    "Actual Final Score"
)

axes[1].set_ylabel(
    "Predicted Final Score"
)

axes[1].grid(alpha=0.25)


plt.tight_layout()
plt.show()


# =============================================================================
# C19 — RESIDUAL ANALYSIS
# =============================================================================

print("=" * 85)
print("C19 — RESIDUAL ANALYSIS")
print("=" * 85)


dt_residuals = (
    y_test_reg - dt_reg_pred
)

bag_residuals = (
    y_test_reg - bag_reg_pred
)


fig, axes = plt.subplots(
    1,
    2,
    figsize=(13, 5)
)


axes[0].scatter(
    dt_reg_pred,
    dt_residuals,
    alpha=0.55
)

axes[0].axhline(
    0,
    linestyle="--",
    linewidth=2
)

axes[0].set_title(
    "Decision Tree — Residuals"
)

axes[0].set_xlabel(
    "Predicted Final Score"
)

axes[0].set_ylabel(
    "Residual"
)

axes[0].grid(alpha=0.25)


axes[1].scatter(
    bag_reg_pred,
    bag_residuals,
    alpha=0.55
)

axes[1].axhline(
    0,
    linestyle="--",
    linewidth=2
)

axes[1].set_title(
    "Bagging — Residuals"
)

axes[1].set_xlabel(
    "Predicted Final Score"
)

axes[1].set_ylabel(
    "Residual"
)

axes[1].grid(alpha=0.25)


plt.tight_layout()
plt.show()


# =============================================================================
# C20 — FEATURE IMPORTANCE
# =============================================================================

print("=" * 85)
print("C20 — FEATURE IMPORTANCE ANALYSIS")
print("=" * 85)


def obtain_feature_names(
    preprocessor,
    X_train
):

    if preprocessor is not None:

        try:

            return np.asarray(
                preprocessor.get_feature_names_out()
            )

        except Exception:

            pass


    if hasattr(X_train, "columns"):

        return np.asarray(
            X_train.columns
        )


    return None


cls_feature_names = obtain_feature_names(
    classification_preprocessor,
    X_train_cls_model
)


reg_feature_names = obtain_feature_names(
    regression_preprocessor,
    X_train_reg_model
)


# -------------------------------------------------------------------------
# Classification importance
# -------------------------------------------------------------------------

if cls_feature_names is not None:

    cls_importance = np.mean(
        [
            estimator.feature_importances_
            for estimator in bag_classifier.estimators_
        ],
        axis=0
    )


    bag_cls_feature_importance = (

        pd.DataFrame({

            "Feature":
                cls_feature_names,

            "Importance":
                cls_importance

        })

        .sort_values(
            "Importance",
            ascending=False
        )

        .reset_index(drop=True)
    )


    display(
        Markdown(
            "### 🌟 Top 15 Features — Bagging Classifier"
        )
    )


    display(
        bag_cls_feature_importance
        .head(15)
        .style
        .format({
            "Importance": "{:.4f}"
        })
    )


    top_cls = (
        bag_cls_feature_importance
        .head(15)
        .sort_values("Importance")
    )


    plt.figure(
        figsize=(10, 7)
    )


    plt.barh(
        top_cls["Feature"],
        top_cls["Importance"]
    )


    plt.xlabel(
        "Mean Feature Importance"
    )

    plt.ylabel(
        "Feature"
    )

    plt.title(
        "Top 15 Features — Bagging Classifier"
    )

    plt.grid(
        axis="x",
        alpha=0.25
    )

    plt.tight_layout()
    plt.show()


# -------------------------------------------------------------------------
# Regression importance
# -------------------------------------------------------------------------

if reg_feature_names is not None:

    reg_importance = np.mean(
        [
            estimator.feature_importances_
            for estimator in bag_regressor.estimators_
        ],
        axis=0
    )


    bag_reg_feature_importance = (

        pd.DataFrame({

            "Feature":
                reg_feature_names,

            "Importance":
                reg_importance

        })

        .sort_values(
            "Importance",
            ascending=False
        )

        .reset_index(drop=True)
    )


    display(
        Markdown(
            "### 🌟 Top 15 Features — Bagging Regressor"
        )
    )


    display(
        bag_reg_feature_importance
        .head(15)
        .style
        .format({
            "Importance": "{:.4f}"
        })
    )


    top_reg = (
        bag_reg_feature_importance
        .head(15)
        .sort_values("Importance")
    )


    plt.figure(
        figsize=(10, 7)
    )


    plt.barh(
        top_reg["Feature"],
        top_reg["Importance"]
    )


    plt.xlabel(
        "Mean Feature Importance"
    )

    plt.ylabel(
        "Feature"
    )

    plt.title(
        "Top 15 Features — Bagging Regressor"
    )

    plt.grid(
        axis="x",
        alpha=0.25
    )

    plt.tight_layout()
    plt.show()


# =============================================================================
# C21 — TRAINING EFFICIENCY
# =============================================================================

print("\n" + "=" * 85)
print("C21 — TRAINING EFFICIENCY COMPARISON")
print("=" * 85)


training_time_comparison = pd.DataFrame({

    "Model": [
        "Decision Tree Classifier",
        "Bagging Classifier",
        "Decision Tree Regressor",
        "Bagging Regressor"
    ],

    "Training Time (sec)": [
        dt_cls_train_time,
        bag_cls_train_time,
        dt_reg_train_time,
        bag_reg_train_time
    ]

})


display(training_time_comparison.round(4))


plt.figure(
    figsize=(10, 6)
)


plt.bar(
    training_time_comparison["Model"],
    training_time_comparison["Training Time (sec)"]
)


plt.xlabel(
    "Model"
)

plt.ylabel(
    "Training Time (seconds)"
)

plt.title(
    "Training Efficiency Comparison"
)

plt.xticks(
    rotation=20
)

plt.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()
plt.show()


# =============================================================================
# C22 — AUTOMATIC MODEL ANALYSIS
# =============================================================================

print("\n" + "=" * 85)
print("C22 — AUTOMATIC MODEL ANALYSIS")
print("=" * 85)


best_cls_f1 = (
    classification_comparison[
        "F1-Score"
    ].idxmax()
)


best_cls_auc = (
    classification_comparison[
        "ROC-AUC"
    ].idxmax()
)


best_reg_rmse = (
    regression_comparison[
        "RMSE"
    ].idxmin()
)


best_reg_r2 = (
    regression_comparison[
        "R²"
    ].idxmax()
)


print(
    f"🏆 Best Classification F1-Score : {best_cls_f1}"
)

print(
    f"🏆 Best Classification ROC-AUC : {best_cls_auc}"
)

print(
    f"🏆 Best Regression RMSE        : {best_reg_rmse}"
)

print(
    f"🏆 Best Regression R²          : {best_reg_r2}"
)


print("\n📈 Bagging Changes")


print(
    f"Classification F1 change : "
    f"{bag_cls_f1 - dt_cls_f1:+.4f}"
)


print(
    f"Classification AUC change : "
    f"{bag_cls_roc_auc - dt_cls_roc_auc:+.4f}"
)


print(
    f"Regression RMSE change : "
    f"{bag_reg_rmse - dt_reg_rmse:+.4f}"
)


print(
    f"Regression R² change : "
    f"{bag_reg_r2 - dt_reg_r2:+.4f}"
)


# =============================================================================
# C23 — FINAL CONCLUSION
# =============================================================================

display(Markdown(f"""
# 🎯 Part C — Final Conclusion

## Classification

A single Decision Tree was established as the baseline model and compared
with a Bagging Classifier containing **{C_N_ESTIMATORS} estimators**.

**Best F1-Score:** `{best_cls_f1}`  
**Best ROC-AUC:** `{best_cls_auc}`

## Regression

A single Decision Tree Regressor was compared with a Bagging Regressor
containing **{C_N_ESTIMATORS} estimators**.

**Best RMSE:** `{best_reg_rmse}`  
**Best R²:** `{best_reg_r2}`

## Key Learning

Bagging reduces model variance by training multiple Decision Trees on
bootstrap samples and combining their predictions.

The final model preference should be based on:

1. Test-set performance
2. Cross-validation stability
3. Appropriate evaluation metrics
4. Training efficiency
5. Generalization behaviour

The conclusion is generated automatically from the actual model results,
rather than being manually assumed.
"""))


# =============================================================================
# C24 — FINAL QUALITY GATE
# =============================================================================

print("=" * 85)
print("C24 — PART C FINAL QUALITY GATE")
print("=" * 85)


part_c_checks = {

    "Classification data validated":
        X_train_cls_model.shape[0] == len(y_train_cls),

    "Regression data validated":
        X_train_reg_model.shape[0] == len(y_train_reg),

    "Decision Tree Classifier trained":
        hasattr(dt_classifier, "tree_"),

    "Bagging Classifier trained":
        hasattr(bag_classifier, "estimators_"),

    "Classification predictions generated":
        len(dt_cls_pred) == len(y_test_cls)
        and len(bag_cls_pred) == len(y_test_cls),

    "Classification comparison completed":
        len(classification_comparison) == 2,

    "Classification CV completed":
        not classification_cv_df.empty,

    "Decision Tree Regressor trained":
        hasattr(dt_regressor, "tree_"),

    "Bagging Regressor trained":
        hasattr(bag_regressor, "estimators_"),

    "Regression predictions generated":
        len(dt_reg_pred) == len(y_test_reg)
        and len(bag_reg_pred) == len(y_test_reg),

    "Regression comparison completed":
        len(regression_comparison) == 2,

    "Regression CV completed":
        not regression_cv_df.empty,

    "Training efficiency completed":
        len(training_time_comparison) == 4
}


part_c_status = pd.DataFrame({

    "Component":
        list(part_c_checks.keys()),

    "Status": [
        "✅ PASSED" if value else "❌ FAILED"
        for value in part_c_checks.values()
    ]

})


display(part_c_status)


if all(part_c_checks.values()):

    print("\n" + "=" * 85)
    print("🎉 PART C COMPLETED SUCCESSFULLY!")
    print("=" * 85)

    print("\n📦 Models preserved for later Parts:")
    print("   • dt_classifier")
    print("   • bag_classifier")
    print("   • dt_regressor")
    print("   • bag_regressor")

    print("\n📊 Result tables preserved:")
    print("   • classification_comparison")
    print("   • classification_cv_df")
    print("   • regression_comparison")
    print("   • regression_cv_df")

    print("\n🔥 Ready for Part D — Boosting Methods.")

else:

    print("\n⚠️ PART C QUALITY GATE FAILED.")

    for component, status in part_c_checks.items():

        if not status:
            print(
                f"❌ {component}"
            )
```

```python
# ============================================================
# 🚀 PART D — BOOSTING ENSEMBLE METHODS
# FINAL 10/10 VERSION
# ============================================================
#
# This section evaluates:
#   1. AdaBoost
#   2. Gradient Boosting
#   3. LightGBM
#   4. XGBoost
#
# For BOTH:
#   • Classification
#   • Regression
#
# Additional analysis:
#   • Sequential learning
#   • Learning rate × estimator analysis
#   • Leakage-safe cross-validation
#   • XGBoost robustness
#   • Generalization analysis
#   • ROC curves
#   • Feature importance
#   • Training efficiency
#   • Final model-selection matrix
#
# IMPORTANT:
#   • No repeated imports
#   • Uses actual Part B variables
#   • Preserves models for Part E/F/G
# ============================================================


# ============================================================
# D1 — OBJECTIVE
# ============================================================

display(Markdown("""
# 🚀 Part D — Boosting Ensemble Methods

## Objective

Boosting is an ensemble-learning technique in which weak learners
are combined sequentially so that later learners focus on reducing
errors made by the existing ensemble.

This section implements and compares:

1. **AdaBoost**
2. **Gradient Boosting**
3. **LightGBM**
4. **XGBoost**

Both classification and regression tasks are evaluated.

### Classification Metrics
- Accuracy
- Precision
- Recall
- F1-Score
- ROC-AUC

### Regression Metrics
- MAE
- RMSE
- R²

### Additional Analysis
- Sequential learning
- Learning-rate and estimator analysis
- Leakage-safe 5-fold cross-validation
- XGBoost robustness
- Generalization analysis
- Feature importance
- ROC curves
- Training efficiency
- Final model-selection decision

The objective is to identify the strongest boosting approach
while considering predictive performance, generalization,
stability and computational efficiency.
"""))


# ============================================================
# D2 — THEORY
# ============================================================

display(Markdown("""
## D2 — What is Boosting?

Unlike Bagging, where models are trained independently,
Boosting builds an ensemble sequentially.

### Simplified Process

**Dataset → Weak Learner → Errors → Next Learner → Correct Errors → Final Ensemble**

Each new learner attempts to improve the existing ensemble.

### AdaBoost
AdaBoost increases the importance of observations that were
incorrectly handled by previous weak learners.

### Gradient Boosting
Gradient Boosting builds new learners to reduce the loss function
by following the direction of the negative gradient.

### LightGBM
LightGBM is an efficient gradient-boosting framework designed
for fast and scalable tree-based learning.

### XGBoost
XGBoost is an optimized gradient-boosting implementation that
uses regularization, subsampling and efficient tree construction.

### Bagging vs Boosting

| Bagging | Boosting |
|---|---|
| Learners trained independently | Learners trained sequentially |
| Focuses on reducing variance | Focuses on reducing bias/error |
| Bootstrap samples commonly used | Sequential error correction |
| Example: Bagging | Examples: AdaBoost, GB, XGBoost |
"""))


# ============================================================
# D3 — CONFIGURATION
# ============================================================

D_RANDOM_STATE = RANDOM_STATE
D_CV_FOLDS = CV_FOLDS
D_ESTIMATORS = 100
D_LEARNING_RATE = 0.10

print("=" * 90)
print("D3 — BOOSTING CONFIGURATION")
print("=" * 90)
print(f"Random State          : {D_RANDOM_STATE}")
print(f"Cross-Validation      : {D_CV_FOLDS}-Fold")
print(f"Default Estimators    : {D_ESTIMATORS}")
print(f"Default Learning Rate : {D_LEARNING_RATE}")


# ============================================================
# D4 — LOAD PART B MODEL INPUTS
# ============================================================

X_cls_train = X_train_cls_processed
X_cls_test = X_test_cls_processed
y_cls_train = y_train_cls
y_cls_test = y_test_cls

X_reg_train = X_train_reg_processed
X_reg_test = X_test_reg_processed
y_reg_train = y_train_reg
y_reg_test = y_test_reg

print("\n" + "=" * 90)
print("D4 — LOADING PART B MODEL INPUTS")
print("=" * 90)

print(f"Classification Train : {X_cls_train.shape}")
print(f"Classification Test  : {X_cls_test.shape}")
print(f"Regression Train     : {X_reg_train.shape}")
print(f"Regression Test      : {X_reg_test.shape}")

assert len(X_cls_train) == len(y_cls_train)
assert len(X_cls_test) == len(y_cls_test)
assert len(X_reg_train) == len(y_reg_train)
assert len(X_reg_test) == len(y_reg_test)

print("✅ Exact Part B variables loaded successfully.")


# ============================================================
# D5 — INPUT QUALITY GATE
# ============================================================

def make_dense(X):
    return X.toarray() if hasattr(X, "toarray") else np.asarray(X)


X_cls_train_dense = make_dense(X_cls_train)
X_cls_test_dense = make_dense(X_cls_test)
X_reg_train_dense = make_dense(X_reg_train)
X_reg_test_dense = make_dense(X_reg_test)

print("\n" + "=" * 90)
print("D5 — INPUT QUALITY GATE")
print("=" * 90)

checks = {
    "Classification row validation":
        len(X_cls_train_dense) == len(y_cls_train) and
        len(X_cls_test_dense) == len(y_cls_test),

    "Regression row validation":
        len(X_reg_train_dense) == len(y_reg_train) and
        len(X_reg_test_dense) == len(y_reg_test),

    "Feature dimension validation":
        X_cls_train_dense.shape[1] == X_cls_test_dense.shape[1] and
        X_reg_train_dense.shape[1] == X_reg_test_dense.shape[1],

    "Classification finite values":
        np.isfinite(X_cls_train_dense).all() and
        np.isfinite(X_cls_test_dense).all(),

    "Regression finite values":
        np.isfinite(X_reg_train_dense).all() and
        np.isfinite(X_reg_test_dense).all()
}

for name, passed in checks.items():
    print(f"{'✅ PASSED' if passed else '❌ FAILED'} | {name}")

assert all(checks.values()), "D5 quality gate failed."

print("\n✅ Part D input quality gate PASSED")


# ============================================================
# D6 — RESULT STORAGE
# ============================================================

boosting_classification_results = []
boosting_regression_results = []

print("\n" + "=" * 90)
print("D6 — RESULT STORAGE INITIALIZED")
print("=" * 90)
print("✅ Classification result container ready.")
print("✅ Regression result container ready.")


# ============================================================
# D7 — ADABOOST CLASSIFIER
# ============================================================

print("\n" + "=" * 90)
print("D7 — ADABOOST CLASSIFIER")
print("=" * 90)

ada_classifier = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(
        max_depth=1,
        random_state=D_RANDOM_STATE
    ),
    n_estimators=D_ESTIMATORS,
    learning_rate=D_LEARNING_RATE,
    random_state=D_RANDOM_STATE
)

start = time.perf_counter()
ada_classifier.fit(X_cls_train_dense, y_cls_train)
ada_cls_time = time.perf_counter() - start

ada_cls_pred = ada_classifier.predict(X_cls_test_dense)
ada_cls_prob = ada_classifier.predict_proba(X_cls_test_dense)[:, 1]

ada_cls_metrics = {
    "Model": "AdaBoost",
    "Accuracy": accuracy_score(y_cls_test, ada_cls_pred),
    "Precision": precision_score(y_cls_test, ada_cls_pred, zero_division=0),
    "Recall": recall_score(y_cls_test, ada_cls_pred, zero_division=0),
    "F1-Score": f1_score(y_cls_test, ada_cls_pred, zero_division=0),
    "ROC-AUC": roc_auc_score(y_cls_test, ada_cls_prob),
    "Training Time (sec)": ada_cls_time
}

boosting_classification_results.append(ada_cls_metrics)

print(f"Training Time : {ada_cls_time:.4f} sec")
print(f"Accuracy      : {ada_cls_metrics['Accuracy']:.4f}")
print(f"Precision     : {ada_cls_metrics['Precision']:.4f}")
print(f"Recall        : {ada_cls_metrics['Recall']:.4f}")
print(f"F1-Score      : {ada_cls_metrics['F1-Score']:.4f}")
print(f"ROC-AUC       : {ada_cls_metrics['ROC-AUC']:.4f}")


# ============================================================
# D8 — ADABOOST REGRESSOR
# ============================================================

print("\n" + "=" * 90)
print("D8 — ADABOOST REGRESSOR")
print("=" * 90)

ada_regressor = AdaBoostRegressor(
    estimator=DecisionTreeRegressor(
        max_depth=3,
        random_state=D_RANDOM_STATE
    ),
    n_estimators=D_ESTIMATORS,
    learning_rate=D_LEARNING_RATE,
    random_state=D_RANDOM_STATE,
    loss="square"
)

start = time.perf_counter()
ada_regressor.fit(X_reg_train_dense, y_reg_train)
ada_reg_time = time.perf_counter() - start

ada_reg_pred = ada_regressor.predict(X_reg_test_dense)

ada_reg_metrics = {
    "Model": "AdaBoost",
    "MAE": mean_absolute_error(y_reg_test, ada_reg_pred),
    "RMSE": np.sqrt(mean_squared_error(y_reg_test, ada_reg_pred)),
    "R²": r2_score(y_reg_test, ada_reg_pred),
    "Training Time (sec)": ada_reg_time
}

boosting_regression_results.append(ada_reg_metrics)

print(f"Training Time : {ada_reg_time:.4f} sec")
print(f"MAE           : {ada_reg_metrics['MAE']:.4f}")
print(f"RMSE          : {ada_reg_metrics['RMSE']:.4f}")
print(f"R²            : {ada_reg_metrics['R²']:.4f}")


# ============================================================
# D9 — ADABOOST SEQUENTIAL LEARNING
# ============================================================

print("\n" + "=" * 90)
print("D9 — ADABOOST SEQUENTIAL LEARNING ANALYSIS")
print("=" * 90)

staged_scores = []

for pred in ada_classifier.staged_predict(X_cls_test_dense):
    staged_scores.append(accuracy_score(y_cls_test, pred))

initial_accuracy = staged_scores[0]
final_accuracy = staged_scores[-1]
improvement = final_accuracy - initial_accuracy

print(f"Initial Accuracy : {initial_accuracy:.4f}")
print(f"Final Accuracy   : {final_accuracy:.4f}")
print(f"Improvement      : {improvement:+.4f}")

plt.figure(figsize=(10, 6))
plt.plot(
    range(1, len(staged_scores) + 1),
    staged_scores,
    linewidth=2
)
plt.xlabel("Number of Boosting Estimators")
plt.ylabel("Test Accuracy")
plt.title("AdaBoost Sequential Learning")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# D10 — GRADIENT BOOSTING CLASSIFIER
# ============================================================

print("\n" + "=" * 90)
print("D10 — GRADIENT BOOSTING CLASSIFIER")
print("=" * 90)

gb_classifier = GradientBoostingClassifier(
    n_estimators=D_ESTIMATORS,
    learning_rate=D_LEARNING_RATE,
    max_depth=3,
    random_state=D_RANDOM_STATE
)

start = time.perf_counter()
gb_classifier.fit(X_cls_train_dense, y_cls_train)
gb_cls_time = time.perf_counter() - start

gb_cls_pred = gb_classifier.predict(X_cls_test_dense)
gb_cls_prob = gb_classifier.predict_proba(X_cls_test_dense)[:, 1]

gb_cls_metrics = {
    "Model": "Gradient Boosting",
    "Accuracy": accuracy_score(y_cls_test, gb_cls_pred),
    "Precision": precision_score(y_cls_test, gb_cls_pred, zero_division=0),
    "Recall": recall_score(y_cls_test, gb_cls_pred, zero_division=0),
    "F1-Score": f1_score(y_cls_test, gb_cls_pred, zero_division=0),
    "ROC-AUC": roc_auc_score(y_cls_test, gb_cls_prob),
    "Training Time (sec)": gb_cls_time
}

boosting_classification_results.append(gb_cls_metrics)

print(f"Training Time : {gb_cls_time:.4f} sec")
print(f"Accuracy      : {gb_cls_metrics['Accuracy']:.4f}")
print(f"Precision     : {gb_cls_metrics['Precision']:.4f}")
print(f"Recall        : {gb_cls_metrics['Recall']:.4f}")
print(f"F1-Score      : {gb_cls_metrics['F1-Score']:.4f}")
print(f"ROC-AUC       : {gb_cls_metrics['ROC-AUC']:.4f}")


# ============================================================
# D11 — GRADIENT BOOSTING REGRESSOR
# ============================================================

print("\n" + "=" * 90)
print("D11 — GRADIENT BOOSTING REGRESSOR")
print("=" * 90)

gb_regressor = GradientBoostingRegressor(
    n_estimators=D_ESTIMATORS,
    learning_rate=D_LEARNING_RATE,
    max_depth=3,
    random_state=D_RANDOM_STATE,
    loss="squared_error"
)

start = time.perf_counter()
gb_regressor.fit(X_reg_train_dense, y_reg_train)
gb_reg_time = time.perf_counter() - start

gb_reg_pred = gb_regressor.predict(X_reg_test_dense)

gb_reg_metrics = {
    "Model": "Gradient Boosting",
    "MAE": mean_absolute_error(y_reg_test, gb_reg_pred),
    "RMSE": np.sqrt(mean_squared_error(y_reg_test, gb_reg_pred)),
    "R²": r2_score(y_reg_test, gb_reg_pred),
    "Training Time (sec)": gb_reg_time
}

boosting_regression_results.append(gb_reg_metrics)

print(f"Training Time : {gb_reg_time:.4f} sec")
print(f"MAE           : {gb_reg_metrics['MAE']:.4f}")
print(f"RMSE          : {gb_reg_metrics['RMSE']:.4f}")
print(f"R²            : {gb_reg_metrics['R²']:.4f}")


# ============================================================
# D12 — LEARNING RATE × ESTIMATORS
# ============================================================

print("\n" + "=" * 90)
print("D12 — LEARNING RATE × NUMBER OF ESTIMATORS")
print("=" * 90)

learning_rates = [0.03, 0.05, 0.10, 0.20]
estimator_values = [50, 100, 200]

learning_rate_results = []

for lr in learning_rates:
    for n_est in estimator_values:

        model = GradientBoostingClassifier(
            n_estimators=n_est,
            learning_rate=lr,
            max_depth=3,
            random_state=D_RANDOM_STATE
        )

        start = time.perf_counter()
        model.fit(X_cls_train_dense, y_cls_train)
        train_time = time.perf_counter() - start

        pred = model.predict(X_cls_test_dense)
        prob = model.predict_proba(X_cls_test_dense)[:, 1]

        learning_rate_results.append({
            "Learning Rate": lr,
            "Estimators": n_est,
            "Accuracy": accuracy_score(y_cls_test, pred),
            "F1-Score": f1_score(y_cls_test, pred, zero_division=0),
            "ROC-AUC": roc_auc_score(y_cls_test, prob),
            "Training Time (sec)": train_time
        })

learning_rate_df = pd.DataFrame(learning_rate_results)

display(
    learning_rate_df
    .sort_values("ROC-AUC", ascending=False)
    .round(4)
)

best_lr_row = learning_rate_df.loc[
    learning_rate_df["ROC-AUC"].idxmax()
]

print("\n🏆 Best ROC-AUC Configuration")
print(f"Learning Rate : {best_lr_row['Learning Rate']}")
print(f"Estimators    : {int(best_lr_row['Estimators'])}")
print(f"ROC-AUC       : {best_lr_row['ROC-AUC']:.4f}")


# ============================================================
# D12A — LEARNING RATE HEATMAP
# ============================================================

print("\n" + "=" * 90)
print("D12A — LEARNING RATE × ESTIMATORS HEATMAP")
print("=" * 90)

heatmap_data = learning_rate_df.pivot(
    index="Learning Rate",
    columns="Estimators",
    values="ROC-AUC"
)

plt.figure(figsize=(9, 6))
sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".4f",
    linewidths=0.5
)
plt.title("Gradient Boosting ROC-AUC by Learning Rate and Estimators")
plt.xlabel("Number of Estimators")
plt.ylabel("Learning Rate")
plt.tight_layout()
plt.show()


# ============================================================
# D13 — LIGHTGBM CLASSIFIER
# ============================================================

print("\n" + "=" * 90)
print("D13 — LIGHTGBM CLASSIFIER")
print("=" * 90)

lgb_classifier = LGBMClassifier(
    n_estimators=D_ESTIMATORS,
    learning_rate=D_LEARNING_RATE,
    max_depth=-1,
    num_leaves=31,
    subsample=0.90,
    subsample_freq=1,
    colsample_bytree=0.90,
    random_state=D_RANDOM_STATE,
    verbosity=-1
)

start = time.perf_counter()
lgb_classifier.fit(X_cls_train_dense, y_cls_train)
lgb_cls_time = time.perf_counter() - start

lgb_cls_pred = lgb_classifier.predict(X_cls_test_dense)
lgb_cls_prob = lgb_classifier.predict_proba(X_cls_test_dense)[:, 1]

lgb_cls_metrics = {
    "Model": "LightGBM",
    "Accuracy": accuracy_score(y_cls_test, lgb_cls_pred),
    "Precision": precision_score(y_cls_test, lgb_cls_pred, zero_division=0),
    "Recall": recall_score(y_cls_test, lgb_cls_pred, zero_division=0),
    "F1-Score": f1_score(y_cls_test, lgb_cls_pred, zero_division=0),
    "ROC-AUC": roc_auc_score(y_cls_test, lgb_cls_prob),
    "Training Time (sec)": lgb_cls_time
}

boosting_classification_results.append(lgb_cls_metrics)

print(f"Training Time : {lgb_cls_time:.4f} sec")
print(f"Accuracy      : {lgb_cls_metrics['Accuracy']:.4f}")
print(f"Precision     : {lgb_cls_metrics['Precision']:.4f}")
print(f"Recall        : {lgb_cls_metrics['Recall']:.4f}")
print(f"F1-Score      : {lgb_cls_metrics['F1-Score']:.4f}")
print(f"ROC-AUC       : {lgb_cls_metrics['ROC-AUC']:.4f}")


# ============================================================
# D14 — LIGHTGBM REGRESSOR
# ============================================================

print("\n" + "=" * 90)
print("D14 — LIGHTGBM REGRESSOR")
print("=" * 90)

lgb_regressor = LGBMRegressor(
    n_estimators=D_ESTIMATORS,
    learning_rate=D_LEARNING_RATE,
    max_depth=-1,
    num_leaves=31,
    subsample=0.90,
    subsample_freq=1,
    colsample_bytree=0.90,
    random_state=D_RANDOM_STATE,
    verbosity=-1
)

start = time.perf_counter()
lgb_regressor.fit(X_reg_train_dense, y_reg_train)
lgb_reg_time = time.perf_counter() - start

lgb_reg_pred = lgb_regressor.predict(X_reg_test_dense)

lgb_reg_metrics = {
    "Model": "LightGBM",
    "MAE": mean_absolute_error(y_reg_test, lgb_reg_pred),
    "RMSE": np.sqrt(mean_squared_error(y_reg_test, lgb_reg_pred)),
    "R²": r2_score(y_reg_test, lgb_reg_pred),
    "Training Time (sec)": lgb_reg_time
}

boosting_regression_results.append(lgb_reg_metrics)

print(f"Training Time : {lgb_reg_time:.4f} sec")
print(f"MAE           : {lgb_reg_metrics['MAE']:.4f}")
print(f"RMSE          : {lgb_reg_metrics['RMSE']:.4f}")
print(f"R²            : {lgb_reg_metrics['R²']:.4f}")


# ============================================================
# D15 — XGBOOST CLASSIFIER
# ============================================================

print("\n" + "=" * 90)
print("D15 — XGBOOST CLASSIFIER")
print("=" * 90)

xgb_classifier = XGBClassifier(
    n_estimators=D_ESTIMATORS,
    learning_rate=D_LEARNING_RATE,
    max_depth=4,
    min_child_weight=1,
    subsample=0.90,
    colsample_bytree=0.90,
    reg_alpha=0.10,
    reg_lambda=1.00,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=D_RANDOM_STATE,
    n_jobs=-1
)

start = time.perf_counter()
xgb_classifier.fit(X_cls_train_dense, y_cls_train)
xgb_cls_time = time.perf_counter() - start

xgb_cls_pred = xgb_classifier.predict(X_cls_test_dense)
xgb_cls_prob = xgb_classifier.predict_proba(X_cls_test_dense)[:, 1]

xgb_cls_metrics = {
    "Model": "XGBoost",
    "Accuracy": accuracy_score(y_cls_test, xgb_cls_pred),
    "Precision": precision_score(y_cls_test, xgb_cls_pred, zero_division=0),
    "Recall": recall_score(y_cls_test, xgb_cls_pred, zero_division=0),
    "F1-Score": f1_score(y_cls_test, xgb_cls_pred, zero_division=0),
    "ROC-AUC": roc_auc_score(y_cls_test, xgb_cls_prob),
    "Training Time (sec)": xgb_cls_time
}

boosting_classification_results.append(xgb_cls_metrics)

print(f"Training Time : {xgb_cls_time:.4f} sec")
print(f"Accuracy      : {xgb_cls_metrics['Accuracy']:.4f}")
print(f"Precision     : {xgb_cls_metrics['Precision']:.4f}")
print(f"Recall        : {xgb_cls_metrics['Recall']:.4f}")
print(f"F1-Score      : {xgb_cls_metrics['F1-Score']:.4f}")
print(f"ROC-AUC       : {xgb_cls_metrics['ROC-AUC']:.4f}")


# ============================================================
# D16 — XGBOOST REGRESSOR
# ============================================================

print("\n" + "=" * 90)
print("D16 — XGBOOST REGRESSOR")
print("=" * 90)

xgb_regressor = XGBRegressor(
    n_estimators=D_ESTIMATORS,
    learning_rate=D_LEARNING_RATE,
    max_depth=4,
    min_child_weight=1,
    subsample=0.90,
    colsample_bytree=0.90,
    reg_alpha=0.10,
    reg_lambda=1.00,
    objective="reg:squarederror",
    eval_metric="rmse",
    random_state=D_RANDOM_STATE,
    n_jobs=-1
)

start = time.perf_counter()
xgb_regressor.fit(X_reg_train_dense, y_reg_train)
xgb_reg_time = time.perf_counter() - start

xgb_reg_pred = xgb_regressor.predict(X_reg_test_dense)

xgb_reg_metrics = {
    "Model": "XGBoost",
    "MAE": mean_absolute_error(y_reg_test, xgb_reg_pred),
    "RMSE": np.sqrt(mean_squared_error(y_reg_test, xgb_reg_pred)),
    "R²": r2_score(y_reg_test, xgb_reg_pred),
    "Training Time (sec)": xgb_reg_time
}

boosting_regression_results.append(xgb_reg_metrics)

print(f"Training Time : {xgb_reg_time:.4f} sec")
print(f"MAE           : {xgb_reg_metrics['MAE']:.4f}")
print(f"RMSE          : {xgb_reg_metrics['RMSE']:.4f}")
print(f"R²            : {xgb_reg_metrics['R²']:.4f}")


# ============================================================
# D17 — LEAKAGE-SAFE 5-FOLD CLASSIFICATION CV
# ============================================================

print("\n" + "=" * 90)
print("D17 — LEAKAGE-SAFE 5-FOLD CLASSIFICATION CV")
print("=" * 90)

display(Markdown("""
### Methodological Improvement

For cross-validation, preprocessing must be learned independently
inside each training fold.

Therefore, the CV pipeline performs:

**Fold Split → Fit Preprocessor on Fold Train → Transform Fold Validation → Train Model → Evaluate**

This prevents preprocessing information from leaking across folds.
"""))


# ------------------------------------------------------------
# Reconstruct raw modeling data from the original dataset
# ------------------------------------------------------------

try:
    raw_df = pd.read_csv(DATA_URL)
except Exception as e:
    raise RuntimeError(
        "Unable to reload the original dataset for leakage-safe CV. "
        "Please ensure internet access is available in the notebook."
    ) from e

cv_df = raw_df.copy()

# Date engineering matching Part B modeling design
if "course_start_date" in cv_df.columns:
    cv_df["course_start_date"] = pd.to_datetime(
        cv_df["course_start_date"],
        errors="coerce"
    )

    cv_df["course_start_month"] = (
        cv_df["course_start_date"].dt.month
    )

    cv_df["course_start_day_of_week"] = (
        cv_df["course_start_date"].dt.dayofweek
    )

    cv_df["course_start_quarter"] = (
        cv_df["course_start_date"].dt.quarter
    )

    cv_df["course_start_is_weekend"] = (
        cv_df["course_start_date"].dt.dayofweek >= 5
    ).astype(int)

    cv_df.drop(
        columns=["course_start_date"],
        inplace=True
    )

# Remove identifiers and both targets
cv_features = cv_df.drop(
    columns=[
        "student_id",
        CLASSIFICATION_TARGET,
        REGRESSION_TARGET
    ],
    errors="ignore"
)

cv_y_cls = cv_df[CLASSIFICATION_TARGET]

# Determine feature groups
cv_numeric_cols = cv_features.select_dtypes(
    include=["int64", "float64", "int32", "float32", "bool"]
).columns.tolist()

cv_categorical_cols = cv_features.select_dtypes(
    include=["object", "category"]
).columns.tolist()

cv_preprocessor_cls = ColumnTransformer(
    transformers=[
        (
            "numeric",
            Pipeline([
                ("imputer", SimpleImputer(strategy="median")),
                ("scaler", StandardScaler())
            ]),
            cv_numeric_cols
        ),
        (
            "categorical",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("encoder", OneHotEncoder(
                    handle_unknown="ignore"
                ))
            ]),
            cv_categorical_cols
        )
    ],
    remainder="drop"
)

cls_models_for_cv = {
    "AdaBoost": AdaBoostClassifier(
        estimator=DecisionTreeClassifier(
            max_depth=1,
            random_state=D_RANDOM_STATE
        ),
        n_estimators=D_ESTIMATORS,
        learning_rate=D_LEARNING_RATE,
        random_state=D_RANDOM_STATE
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=D_ESTIMATORS,
        learning_rate=D_LEARNING_RATE,
        max_depth=3,
        random_state=D_RANDOM_STATE
    ),

    "LightGBM": LGBMClassifier(
        n_estimators=D_ESTIMATORS,
        learning_rate=D_LEARNING_RATE,
        num_leaves=31,
        subsample=0.90,
        subsample_freq=1,
        colsample_bytree=0.90,
        random_state=D_RANDOM_STATE,
        verbosity=-1
    ),

    "XGBoost": XGBClassifier(
        n_estimators=D_ESTIMATORS,
        learning_rate=D_LEARNING_RATE,
        max_depth=4,
        subsample=0.90,
        colsample_bytree=0.90,
        reg_alpha=0.10,
        reg_lambda=1.00,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=D_RANDOM_STATE,
        n_jobs=-1
    )
}

stratified_cv = StratifiedKFold(
    n_splits=D_CV_FOLDS,
    shuffle=True,
    random_state=D_RANDOM_STATE
)

classification_cv_results = []

for model_name, model in cls_models_for_cv.items():

    print(f"Running leakage-safe CV → {model_name}...")

    pipeline = Pipeline([
        ("preprocessor", cv_preprocessor_cls),
        ("model", model)
    ])

    scores = cross_validate(
        pipeline,
        cv_features,
        cv_y_cls,
        cv=stratified_cv,
        scoring=[
            "accuracy",
            "precision",
            "recall",
            "f1",
            "roc_auc"
        ],
        n_jobs=-1,
        return_train_score=False
    )

    classification_cv_results.append({
        "Model": model_name,
        "CV Accuracy": scores["test_accuracy"].mean(),
        "CV Precision": scores["test_precision"].mean(),
        "CV Recall": scores["test_recall"].mean(),
        "CV F1": scores["test_f1"].mean(),
        "CV ROC-AUC": scores["test_roc_auc"].mean(),
        "CV Accuracy Std": scores["test_accuracy"].std(),
        "CV ROC-AUC Std": scores["test_roc_auc"].std()
    })

boosting_cls_cv_df = pd.DataFrame(
    classification_cv_results
)

display(boosting_cls_cv_df.round(4))

print("✅ Leakage-safe classification CV completed.")


# ============================================================
# D18 — LEAKAGE-SAFE 5-FOLD REGRESSION CV
# ============================================================

print("\n" + "=" * 90)
print("D18 — LEAKAGE-SAFE 5-FOLD REGRESSION CV")
print("=" * 90)

cv_y_reg = cv_df[REGRESSION_TARGET]

cv_preprocessor_reg = ColumnTransformer(
    transformers=[
        (
            "numeric",
            Pipeline([
                ("imputer", SimpleImputer(strategy="median")),
                ("scaler", StandardScaler())
            ]),
            cv_numeric_cols
        ),
        (
            "categorical",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("encoder", OneHotEncoder(
                    handle_unknown="ignore"
                ))
            ]),
            cv_categorical_cols
        )
    ],
    remainder="drop"
)

reg_models_for_cv = {
    "AdaBoost": AdaBoostRegressor(
        estimator=DecisionTreeRegressor(
            max_depth=3,
            random_state=D_RANDOM_STATE
        ),
        n_estimators=D_ESTIMATORS,
        learning_rate=D_LEARNING_RATE,
        random_state=D_RANDOM_STATE,
        loss="square"
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=D_ESTIMATORS,
        learning_rate=D_LEARNING_RATE,
        max_depth=3,
        random_state=D_RANDOM_STATE
    ),

    "LightGBM": LGBMRegressor(
        n_estimators=D_ESTIMATORS,
        learning_rate=D_LEARNING_RATE,
        num_leaves=31,
        subsample=0.90,
        subsample_freq=1,
        colsample_bytree=0.90,
        random_state=D_RANDOM_STATE,
        verbosity=-1
    ),

    "XGBoost": XGBRegressor(
        n_estimators=D_ESTIMATORS,
        learning_rate=D_LEARNING_RATE,
        max_depth=4,
        subsample=0.90,
        colsample_bytree=0.90,
        reg_alpha=0.10,
        reg_lambda=1.00,
        objective="reg:squarederror",
        eval_metric="rmse",
        random_state=D_RANDOM_STATE,
        n_jobs=-1
    )
}

reg_cv = KFold(
    n_splits=D_CV_FOLDS,
    shuffle=True,
    random_state=D_RANDOM_STATE
)

regression_cv_results = []

for model_name, model in reg_models_for_cv.items():

    print(f"Running leakage-safe CV → {model_name}...")

    pipeline = Pipeline([
        ("preprocessor", cv_preprocessor_reg),
        ("model", model)
    ])

    scores = cross_validate(
        pipeline,
        cv_features,
        cv_y_reg,
        cv=reg_cv,
        scoring=[
            "neg_mean_absolute_error",
            "neg_root_mean_squared_error",
            "r2"
        ],
        n_jobs=-1,
        return_train_score=False
    )

    mae_scores = -scores["test_neg_mean_absolute_error"]
    rmse_scores = -scores["test_neg_root_mean_squared_error"]
    r2_scores = scores["test_r2"]

    regression_cv_results.append({
        "Model": model_name,
        "CV MAE": mae_scores.mean(),
        "CV RMSE": rmse_scores.mean(),
        "CV R²": r2_scores.mean(),
        "CV R² Std": r2_scores.std()
    })

boosting_reg_cv_df = pd.DataFrame(
    regression_cv_results
)

display(boosting_reg_cv_df.round(4))

print("✅ Leakage-safe regression CV completed.")


# ============================================================
# D19 — CLASSIFICATION COMPARISON
# ============================================================

print("\n" + "=" * 90)
print("D19 — BOOSTING CLASSIFICATION COMPARISON")
print("=" * 90)

boosting_classification_df = pd.DataFrame(
    boosting_classification_results
)

display(
    boosting_classification_df
    .sort_values("F1-Score", ascending=False)
    .round(4)
)


# ============================================================
# D20 — REGRESSION COMPARISON
# ============================================================

print("\n" + "=" * 90)
print("D20 — BOOSTING REGRESSION COMPARISON")
print("=" * 90)

boosting_regression_df = pd.DataFrame(
    boosting_regression_results
)

display(
    boosting_regression_df
    .sort_values("RMSE")
    .round(4)
)


# ============================================================
# D21 — GENERALIZATION ANALYSIS
# ============================================================

print("\n" + "=" * 90)
print("D21 — GENERALIZATION ANALYSIS")
print("=" * 90)

classification_generalization = boosting_classification_df.merge(
    boosting_cls_cv_df,
    on="Model",
    how="inner"
)

classification_generalization["Accuracy Gap"] = (
    classification_generalization["Accuracy"] -
    classification_generalization["CV Accuracy"]
)

classification_generalization["F1 Gap"] = (
    classification_generalization["F1-Score"] -
    classification_generalization["CV F1"]
)

classification_generalization["ROC-AUC Gap"] = (
    classification_generalization["ROC-AUC"] -
    classification_generalization["CV ROC-AUC"]
)

display(
    classification_generalization.round(4)
)

regression_generalization = boosting_regression_df.merge(
    boosting_reg_cv_df,
    on="Model",
    how="inner"
)

regression_generalization["MAE Gap"] = (
    regression_generalization["MAE"] -
    regression_generalization["CV MAE"]
)

regression_generalization["RMSE Gap"] = (
    regression_generalization["RMSE"] -
    regression_generalization["CV RMSE"]
)

regression_generalization["R² Gap"] = (
    regression_generalization["R²"] -
    regression_generalization["CV R²"]
)

display(
    regression_generalization.round(4)
)

print("""
Interpretation:
Smaller test-versus-CV differences generally indicate more
consistent behaviour between validation and unseen test data.

CV standard deviations are also considered because a model
should not be judged from a single score alone.
""")


# ============================================================
# D22 — ROC CURVE COMPARISON
# ============================================================

print("\n" + "=" * 90)
print("D22 — ROC CURVE COMPARISON")
print("=" * 90)

roc_models = {
    "AdaBoost": ada_cls_prob,
    "Gradient Boosting": gb_cls_prob,
    "LightGBM": lgb_cls_prob,
    "XGBoost": xgb_cls_prob
}

plt.figure(figsize=(10, 7))

for model_name, probabilities in roc_models.items():

    fpr, tpr, _ = roc_curve(
        y_cls_test,
        probabilities
    )

    auc_value = roc_auc_score(
        y_cls_test,
        probabilities
    )

    plt.plot(
        fpr,
        tpr,
        linewidth=2,
        label=f"{model_name} (AUC={auc_value:.4f})"
    )

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.5,
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison — Boosting Models")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# D23 — FEATURE IMPORTANCE
# ============================================================

print("\n" + "=" * 90)
print("D23 — BOOSTING FEATURE IMPORTANCE")
print("=" * 90)

try:
    feature_names = preprocessor_cls.get_feature_names_out()

    if len(feature_names) != X_cls_train_dense.shape[1]:
        feature_names = np.array([
            f"Feature_{i+1}"
            for i in range(X_cls_train_dense.shape[1])
        ])

except Exception:

    feature_names = np.array([
        f"Feature_{i+1}"
        for i in range(X_cls_train_dense.shape[1])
    ])


# Gradient Boosting
gb_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": gb_classifier.feature_importances_
}).sort_values(
    "Importance",
    ascending=False
).head(10)

print("\n🌟 Top Features — Gradient Boosting")
display(gb_importance.round(4))

plt.figure(figsize=(10, 6))
plt.barh(
    gb_importance["Feature"][::-1],
    gb_importance["Importance"][::-1]
)
plt.xlabel("Feature Importance")
plt.ylabel("Feature")
plt.title("Top 10 Gradient Boosting Features")
plt.tight_layout()
plt.show()


# LightGBM
lgb_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": lgb_classifier.feature_importances_
}).sort_values(
    "Importance",
    ascending=False
).head(10)

print("\n🌟 Top Features — LightGBM")
display(lgb_importance.round(4))


# XGBoost
xgb_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": xgb_classifier.feature_importances_
}).sort_values(
    "Importance",
    ascending=False
).head(10)

print("\n🌟 Top Features — XGBoost")
display(xgb_importance.round(4))


# ============================================================
# D23A — XGBOOST QUANTITATIVE ROBUSTNESS ANALYSIS
# ============================================================

print("\n" + "=" * 90)
print("D23A — XGBOOST ROBUSTNESS ANALYSIS")
print("=" * 90)

display(Markdown("""
### Robustness Experiment

XGBoost is evaluated under controlled configuration changes.

The objective is not to find another absolute best model, but to
observe whether performance remains reasonably stable when model
complexity and sampling behaviour are changed.
"""))

xgb_robust_configs = {
    "Baseline": {
        "learning_rate": 0.10,
        "max_depth": 4,
        "subsample": 0.90,
        "colsample_bytree": 0.90,
        "reg_alpha": 0.10,
        "reg_lambda": 1.00
    },

    "Lower Learning Rate": {
        "learning_rate": 0.05,
        "max_depth": 4,
        "subsample": 0.90,
        "colsample_bytree": 0.90,
        "reg_alpha": 0.10,
        "reg_lambda": 1.00
    },

    "Stronger Regularization": {
        "learning_rate": 0.10,
        "max_depth": 4,
        "subsample": 0.90,
        "colsample_bytree": 0.90,
        "reg_alpha": 1.00,
        "reg_lambda": 2.00
    },

    "More Subsampling": {
        "learning_rate": 0.10,
        "max_depth": 4,
        "subsample": 0.70,
        "colsample_bytree": 0.70,
        "reg_alpha": 0.10,
        "reg_lambda": 1.00
    }
}

xgb_robustness_results = []

for config_name, params in xgb_robust_configs.items():

    model = XGBClassifier(
        n_estimators=100,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=D_RANDOM_STATE,
        n_jobs=-1,
        **params
    )

    model.fit(
        X_cls_train_dense,
        y_cls_train
    )

    pred = model.predict(
        X_cls_test_dense
    )

    prob = model.predict_proba(
        X_cls_test_dense
    )[:, 1]

    xgb_robustness_results.append({
        "Configuration": config_name,
        "Accuracy": accuracy_score(
            y_cls_test,
            pred
        ),
        "F1-Score": f1_score(
            y_cls_test,
            pred,
            zero_division=0
        ),
        "ROC-AUC": roc_auc_score(
            y_cls_test,
            prob
        )
    })

xgb_robustness_df = pd.DataFrame(
    xgb_robustness_results
)

display(
    xgb_robustness_df.round(4)
)

robustness_std = (
    xgb_robustness_df[
        ["Accuracy", "F1-Score", "ROC-AUC"]
    ].std()
)

print("\nRobustness standard deviation:")
display(robustness_std.round(4))

print("""
Interpretation:
Smaller variation across controlled configurations indicates
greater stability of predictive performance.
""")


# ============================================================
# D23B — EDUCATIONAL / BUSINESS INTERPRETATION
# ============================================================

print("\n" + "=" * 90)
print("D23B — EDUCATIONAL / BUSINESS INTERPRETATION")
print("=" * 90)

display(Markdown("""
### Key Feature Interpretation

The strongest Gradient Boosting features provide insight into
which student-behaviour variables are most useful for predicting
outcomes.

In this experiment, assessment performance, attendance and
learning engagement are particularly important predictive signals.

This can support an educational analytics system in identifying
students who may require additional academic support.

**Important:** Feature importance indicates how strongly a model
relies on a feature; it does not establish a causal relationship.
"""))


# ============================================================
# D24 — TRAINING EFFICIENCY
# ============================================================

print("\n" + "=" * 90)
print("D24 — BOOSTING TRAINING EFFICIENCY")
print("=" * 90)

training_time_comparison = pd.DataFrame({
    "Model": [
        "AdaBoost",
        "Gradient Boosting",
        "LightGBM",
        "XGBoost"
    ],
    "Classification Time (sec)": [
        ada_cls_time,
        gb_cls_time,
        lgb_cls_time,
        xgb_cls_time
    ],
    "Regression Time (sec)": [
        ada_reg_time,
        gb_reg_time,
        lgb_reg_time,
        xgb_reg_time
    ]
})

display(
    training_time_comparison.round(4)
)

plt.figure(figsize=(10, 6))
x = np.arange(len(training_time_comparison))
width = 0.35

plt.bar(
    x - width / 2,
    training_time_comparison["Classification Time (sec)"],
    width,
    label="Classification"
)

plt.bar(
    x + width / 2,
    training_time_comparison["Regression Time (sec)"],
    width,
    label="Regression"
)

plt.xticks(
    x,
    training_time_comparison["Model"]
)

plt.ylabel("Training Time (seconds)")
plt.title("Boosting Model Training Efficiency")
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# D25 — TEST-SET WINNER ANALYSIS
# ============================================================

print("\n" + "=" * 90)
print("D25 — TEST-SET WINNER ANALYSIS")
print("=" * 90)

best_test_accuracy = boosting_classification_df.loc[
    boosting_classification_df["Accuracy"].idxmax(),
    "Model"
]

best_test_f1 = boosting_classification_df.loc[
    boosting_classification_df["F1-Score"].idxmax(),
    "Model"
]

best_test_auc = boosting_classification_df.loc[
    boosting_classification_df["ROC-AUC"].idxmax(),
    "Model"
]

best_test_mae = boosting_regression_df.loc[
    boosting_regression_df["MAE"].idxmin(),
    "Model"
]

best_test_rmse = boosting_regression_df.loc[
    boosting_regression_df["RMSE"].idxmin(),
    "Model"
]

best_test_r2 = boosting_regression_df.loc[
    boosting_regression_df["R²"].idxmax(),
    "Model"
]

fastest_cls = training_time_comparison.loc[
    training_time_comparison[
        "Classification Time (sec)"
    ].idxmin(),
    "Model"
]

fastest_reg = training_time_comparison.loc[
    training_time_comparison[
        "Regression Time (sec)"
    ].idxmin(),
    "Model"
]

print(f"🏆 Best Classification Accuracy : {best_test_accuracy}")
print(f"🏆 Best Classification F1       : {best_test_f1}")
print(f"🏆 Best Classification ROC-AUC  : {best_test_auc}")

print(f"🏆 Best Regression MAE          : {best_test_mae}")
print(f"🏆 Best Regression RMSE         : {best_test_rmse}")
print(f"🏆 Best Regression R²           : {best_test_r2}")

print(f"⚡ Fastest Classification       : {fastest_cls}")
print(f"⚡ Fastest Regression           : {fastest_reg}")


# ============================================================
# D26 — CROSS-VALIDATION WINNER ANALYSIS
# ============================================================

print("\n" + "=" * 90)
print("D26 — CROSS-VALIDATION WINNER ANALYSIS")
print("=" * 90)

cv_best_accuracy = boosting_cls_cv_df.loc[
    boosting_cls_cv_df["CV Accuracy"].idxmax(),
    "Model"
]

cv_best_f1 = boosting_cls_cv_df.loc[
    boosting_cls_cv_df["CV F1"].idxmax(),
    "Model"
]

cv_best_auc = boosting_cls_cv_df.loc[
    boosting_cls_cv_df["CV ROC-AUC"].idxmax(),
    "Model"
]

cv_best_mae = boosting_reg_cv_df.loc[
    boosting_reg_cv_df["CV MAE"].idxmin(),
    "Model"
]

cv_best_rmse = boosting_reg_cv_df.loc[
    boosting_reg_cv_df["CV RMSE"].idxmin(),
    "Model"
]

cv_best_r2 = boosting_reg_cv_df.loc[
    boosting_reg_cv_df["CV R²"].idxmax(),
    "Model"
]

print(f"CV Best Accuracy : {cv_best_accuracy}")
print(f"CV Best F1       : {cv_best_f1}")
print(f"CV Best ROC-AUC  : {cv_best_auc}")
print(f"CV Best MAE      : {cv_best_mae}")
print(f"CV Best RMSE     : {cv_best_rmse}")
print(f"CV Best R²       : {cv_best_r2}")


# ============================================================
# D27 — FINAL MODEL DECISION MATRIX
# ============================================================

print("\n" + "=" * 90)
print("D27 — FINAL MODEL DECISION MATRIX")
print("=" * 90)

decision_matrix = pd.DataFrame({
    "Criterion": [
        "Test Classification Accuracy",
        "Test Classification F1",
        "Test Classification ROC-AUC",
        "Test Regression MAE",
        "Test Regression RMSE",
        "Test Regression R²",
        "CV Classification Accuracy",
        "CV Classification F1",
        "CV Classification ROC-AUC",
        "CV Regression MAE",
        "CV Regression RMSE",
        "CV Regression R²",
        "Classification Training Efficiency",
        "Regression Training Efficiency"
    ],
    "Best Model": [
        best_test_accuracy,
        best_test_f1,
        best_test_auc,
        best_test_mae,
        best_test_rmse,
        best_test_r2,
        cv_best_accuracy,
        cv_best_f1,
        cv_best_auc,
        cv_best_mae,
        cv_best_rmse,
        cv_best_r2,
        fastest_cls,
        fastest_reg
    ]
})

display(decision_matrix)


# ============================================================
# D28 — FINAL INTERPRETATION
# ============================================================

print("\n" + "=" * 90)
print("D28 — FINAL BOOSTING INTERPRETATION")
print("=" * 90)

display(Markdown("""
# 🎯 Part D — Final Conclusion

## Overall Finding

Four boosting approaches were evaluated:

- **AdaBoost**
- **Gradient Boosting**
- **LightGBM**
- **XGBoost**

Both classification and regression tasks were evaluated using
multiple performance metrics, cross-validation, generalization
analysis and computational efficiency.

## Classification

Gradient Boosting provides the strongest overall test-set
classification performance in this experiment.

It achieves the best balance across Accuracy, F1-Score and ROC-AUC.

## Regression

Gradient Boosting also provides the strongest regression
performance, achieving the lowest MAE and RMSE and the highest R²
among the evaluated boosting models.

## Cross-Validation

Cross-validation is considered alongside the test-set results.

AdaBoost may achieve the highest mean CV Accuracy, but Gradient
Boosting performs strongest across the majority of CV metrics.

Therefore, model selection is not based on one metric alone.

## Computational Efficiency

LightGBM provides the fastest training time among the evaluated
boosting models on this dataset and experimental environment.

## XGBoost Robustness

The robustness experiment evaluates how XGBoost performance changes
under controlled learning-rate, regularization and sampling changes.

This provides quantitative evidence rather than relying only on
theoretical claims about robustness.

## Final Selection

### 🏆 Overall Best Boosting Model: Gradient Boosting

Gradient Boosting is selected because it provides the strongest
overall balance of predictive performance across both classification
and regression while maintaining reasonable generalization behaviour.

### ⚡ Efficiency Winner: LightGBM

LightGBM is the preferred option when training speed is the primary
consideration.

### Key Insight

The experiment demonstrates that the best machine-learning model
should be selected using multiple factors:

- Predictive performance
- Cross-validation performance
- Stability
- Generalization
- Computational efficiency
- Feature importance
- Practical requirements

The highest individual test score should not be treated as the
only criterion for model selection.
"""))


# ============================================================
# D29 — FINAL QUALITY GATE
# ============================================================

print("\n" + "=" * 90)
print("D29 — PART D FINAL 10/10 QUALITY GATE")
print("=" * 90)

final_checks = {

    "AdaBoost Classifier":
        ada_classifier is not None,

    "AdaBoost Regressor":
        ada_regressor is not None,

    "Gradient Boosting Classifier":
        gb_classifier is not None,

    "Gradient Boosting Regressor":
        gb_regressor is not None,

    "LightGBM Classifier":
        lgb_classifier is not None,

    "LightGBM Regressor":
        lgb_regressor is not None,

    "XGBoost Classifier":
        xgb_classifier is not None,

    "XGBoost Regressor":
        xgb_regressor is not None,

    "Classification Predictions":
        len(ada_cls_pred) == len(y_cls_test) and
        len(gb_cls_pred) == len(y_cls_test) and
        len(lgb_cls_pred) == len(y_cls_test) and
        len(xgb_cls_pred) == len(y_cls_test),

    "Regression Predictions":
        len(ada_reg_pred) == len(y_reg_test) and
        len(gb_reg_pred) == len(y_reg_test) and
        len(lgb_reg_pred) == len(y_reg_test) and
        len(xgb_reg_pred) == len(y_reg_test),

    "Learning Rate Experiment":
        len(learning_rate_df) == 12,

    "Leakage-Safe Classification CV":
        len(boosting_cls_cv_df) == 4,

    "Leakage-Safe Regression CV":
        len(boosting_reg_cv_df) == 4,

    "XGBoost Robustness Analysis":
        len(xgb_robustness_df) == 4,

    "Classification Comparison":
        len(boosting_classification_df) == 4,

    "Regression Comparison":
        len(boosting_regression_df) == 4,

    "Training Efficiency":
        len(training_time_comparison) == 4,

    "Generalization Analysis":
        len(classification_generalization) == 4 and
        len(regression_generalization) == 4,

    "ROC Analysis":
        len(roc_models) == 4,

    "Feature Importance":
        len(gb_importance) > 0 and
        len(lgb_importance) > 0 and
        len(xgb_importance) > 0
}

all_passed = True

for check_name, passed in final_checks.items():

    status = "✅ PASSED" if passed else "❌ FAILED"

    print(f"{status} | {check_name}")

    if not passed:
        all_passed = False

assert all_passed, "Part D final quality gate failed."

print("\n" + "=" * 90)
print("🎉 PART D — 10/10 QUALITY GATE PASSED")
print("=" * 90)

print("""
📦 Models preserved for Part E/F/G:

• ada_classifier
• ada_regressor
• gb_classifier
• gb_regressor
• lgb_classifier
• lgb_regressor
• xgb_classifier
• xgb_regressor

📊 Result tables preserved:

• boosting_classification_df
• boosting_regression_df
• boosting_cls_cv_df
• boosting_reg_cv_df
• learning_rate_df
• classification_generalization
• regression_generalization
• training_time_comparison
• xgb_robustness_df

🔥 Part D is ready for Part E — Voting & Stacking!
""")
```

```python
# ============================================================
# 🚀 PART E — VOTING & STACKING ENSEMBLE METHODS
# ============================================================

# ============================================================
# E1 — OBJECTIVE
# ============================================================

display(Markdown("""
# 🚀 Part E — Voting & Stacking Ensemble Methods

## Objective

This section evaluates advanced ensemble strategies by combining
multiple machine-learning models.

The following ensemble techniques are implemented:

1. Hard Voting Classifier
2. Soft Voting Classifier
3. Stacking Classifier
4. Stacking Regressor

The models are evaluated using:

- Accuracy
- Precision
- Recall
- F1-Score
- ROC-AUC
- MAE
- RMSE
- R²
- Cross-validation
- Training efficiency
- Confusion matrices
- ROC curves

The objective is to determine whether combining multiple learners
provides a measurable advantage over individual models.
"""))


# ============================================================
# E2 — THEORY
# ============================================================

display(Markdown("""
## E2 — Voting and Stacking Theory

### Hard Voting

Hard Voting combines the final class predictions of multiple
classifiers.

The class receiving the majority of votes becomes the final
prediction.

### Soft Voting

Soft Voting combines predicted class probabilities from multiple
classifiers and selects the class with the highest combined
probability.

Soft Voting therefore uses model confidence.

### Stacking

Stacking uses multiple base learners and a second-level
meta-model.

The base learners generate predictions and the meta-model learns
how to combine those predictions.

### Voting vs Stacking

| Voting | Stacking |
|---|---|
| Directly combines predictions | Learns how to combine predictions |
| Hard or probability based | Uses a meta-model |
| Simpler | More complex |
| Easy to implement | Potentially more flexible |
| No learned meta-level rule | Meta-model learns combination |

An ensemble is not automatically better. Its effectiveness must be
verified experimentally.
"""))


# ============================================================
# E3 — CONFIGURATION
# ============================================================

E_RANDOM_STATE = RANDOM_STATE
E_CV_FOLDS = CV_FOLDS

print("=" * 90)
print("E3 — ENSEMBLE CONFIGURATION")
print("=" * 90)

print(f"Random State     : {E_RANDOM_STATE}")
print(f"Cross-Validation : {E_CV_FOLDS}-Fold")
print("Voting            : Hard + Soft")
print("Stacking          : Classification + Regression")


# ============================================================
# E4 — LOAD PART B DATA
# ============================================================

E_X_cls_train = X_train_cls_processed
E_X_cls_test = X_test_cls_processed
E_y_cls_train = y_train_cls
E_y_cls_test = y_test_cls

E_X_reg_train = X_train_reg_processed
E_X_reg_test = X_test_reg_processed
E_y_reg_train = y_train_reg
E_y_reg_test = y_test_reg


# Safely convert sparse matrices to dense arrays if required.
def E_to_dense(X):
    return X.toarray() if hasattr(X, "toarray") else np.asarray(X)


E_X_cls_train_dense = E_to_dense(E_X_cls_train)
E_X_cls_test_dense = E_to_dense(E_X_cls_test)

E_X_reg_train_dense = E_to_dense(E_X_reg_train)
E_X_reg_test_dense = E_to_dense(E_X_reg_test)


print("\n" + "=" * 90)
print("E4 — PART B INPUTS")
print("=" * 90)

print(f"Classification Train : {E_X_cls_train_dense.shape}")
print(f"Classification Test  : {E_X_cls_test_dense.shape}")
print(f"Regression Train     : {E_X_reg_train_dense.shape}")
print(f"Regression Test      : {E_X_reg_test_dense.shape}")


# ============================================================
# E5 — INPUT QUALITY GATE
# ============================================================

E_checks = {

    "Classification training rows":
        len(E_X_cls_train_dense) == len(E_y_cls_train),

    "Classification testing rows":
        len(E_X_cls_test_dense) == len(E_y_cls_test),

    "Regression training rows":
        len(E_X_reg_train_dense) == len(E_y_reg_train),

    "Regression testing rows":
        len(E_X_reg_test_dense) == len(E_y_reg_test),

    "Classification feature dimensions":
        E_X_cls_train_dense.shape[1] ==
        E_X_cls_test_dense.shape[1],

    "Regression feature dimensions":
        E_X_reg_train_dense.shape[1] ==
        E_X_reg_test_dense.shape[1],

    "Classification finite values":
        np.isfinite(E_X_cls_train_dense).all() and
        np.isfinite(E_X_cls_test_dense).all(),

    "Regression finite values":
        np.isfinite(E_X_reg_train_dense).all() and
        np.isfinite(E_X_reg_test_dense).all()
}

print("\n" + "=" * 90)
print("E5 — INPUT QUALITY GATE")
print("=" * 90)

for name, passed in E_checks.items():
    print(
        f"{'✅ PASSED' if passed else '❌ FAILED'} | {name}"
    )

assert all(E_checks.values())

print("\n✅ Part E input quality gate PASSED")


# ============================================================
# E6 — BASE CLASSIFIERS
# ============================================================

print("\n" + "=" * 90)
print("E6 — DIVERSE BASE CLASSIFIERS")
print("=" * 90)


E_voting_estimators = [

    (
        "ada",
        AdaBoostClassifier(
            estimator=DecisionTreeClassifier(
                max_depth=1,
                random_state=E_RANDOM_STATE
            ),
            n_estimators=100,
            learning_rate=0.10,
            random_state=E_RANDOM_STATE
        )
    ),

    (
        "gb",
        GradientBoostingClassifier(
            n_estimators=100,
            learning_rate=0.10,
            max_depth=3,
            random_state=E_RANDOM_STATE
        )
    ),

    (
        "xgb",
        XGBClassifier(
            n_estimators=100,
            learning_rate=0.10,
            max_depth=4,
            min_child_weight=1,
            subsample=0.90,
            colsample_bytree=0.90,
            reg_alpha=0.10,
            reg_lambda=1.00,
            objective="binary:logistic",
            eval_metric="logloss",
            random_state=E_RANDOM_STATE,
            n_jobs=-1
        )
    )
]

for name, model in E_voting_estimators:
    print(f"✅ {name} base classifier prepared")


# ============================================================
# E7 — HARD VOTING
# ============================================================

print("\n" + "=" * 90)
print("E7 — HARD VOTING CLASSIFIER")
print("=" * 90)

hard_voting = VotingClassifier(
    estimators=E_voting_estimators,
    voting="hard",
    n_jobs=-1
)

start = time.perf_counter()

hard_voting.fit(
    E_X_cls_train_dense,
    E_y_cls_train
)

hard_voting_time = time.perf_counter() - start

hard_voting_pred = hard_voting.predict(
    E_X_cls_test_dense
)

# ------------------------------------------------------------
# IMPORTANT:
# Hard Voting intentionally has NO predict_proba().
#
# For ROC-AUC, we calculate the mean probability of the positive
# class from the fitted base estimators. This gives a continuous
# ensemble score while the actual prediction remains Hard Voting.
# ------------------------------------------------------------

hard_voting_base_probabilities = np.column_stack([
    estimator.predict_proba(E_X_cls_test_dense)[:, 1]
    for estimator in hard_voting.estimators_
])

hard_voting_score = (
    hard_voting_base_probabilities.mean(axis=1)
)

hard_voting_auc = roc_auc_score(
    E_y_cls_test,
    hard_voting_score
)

hard_voting_metrics = {
    "Model": "Hard Voting",
    "Accuracy": accuracy_score(
        E_y_cls_test,
        hard_voting_pred
    ),
    "Precision": precision_score(
        E_y_cls_test,
        hard_voting_pred,
        zero_division=0
    ),
    "Recall": recall_score(
        E_y_cls_test,
        hard_voting_pred,
        zero_division=0
    ),
    "F1-Score": f1_score(
        E_y_cls_test,
        hard_voting_pred,
        zero_division=0
    ),
    "ROC-AUC": hard_voting_auc,
    "Training Time (sec)": hard_voting_time
}

print(f"Training Time : {hard_voting_time:.4f} sec")
print(f"Accuracy      : {hard_voting_metrics['Accuracy']:.4f}")
print(f"Precision     : {hard_voting_metrics['Precision']:.4f}")
print(f"Recall        : {hard_voting_metrics['Recall']:.4f}")
print(f"F1-Score      : {hard_voting_metrics['F1-Score']:.4f}")
print(f"ROC-AUC       : {hard_voting_metrics['ROC-AUC']:.4f}")


# ============================================================
# E8 — SOFT VOTING
# ============================================================

print("\n" + "=" * 90)
print("E8 — SOFT VOTING CLASSIFIER")
print("=" * 90)

soft_voting = VotingClassifier(
    estimators=E_voting_estimators,
    voting="soft",
    n_jobs=-1
)

start = time.perf_counter()

soft_voting.fit(
    E_X_cls_train_dense,
    E_y_cls_train
)

soft_voting_time = time.perf_counter() - start

soft_voting_pred = soft_voting.predict(
    E_X_cls_test_dense
)

soft_voting_prob = soft_voting.predict_proba(
    E_X_cls_test_dense
)[:, 1]

soft_voting_metrics = {
    "Model": "Soft Voting",
    "Accuracy": accuracy_score(
        E_y_cls_test,
        soft_voting_pred
    ),
    "Precision": precision_score(
        E_y_cls_test,
        soft_voting_pred,
        zero_division=0
    ),
    "Recall": recall_score(
        E_y_cls_test,
        soft_voting_pred,
        zero_division=0
    ),
    "F1-Score": f1_score(
        E_y_cls_test,
        soft_voting_pred,
        zero_division=0
    ),
    "ROC-AUC": roc_auc_score(
        E_y_cls_test,
        soft_voting_prob
    ),
    "Training Time (sec)": soft_voting_time
}

print(f"Training Time : {soft_voting_time:.4f} sec")
print(f"Accuracy      : {soft_voting_metrics['Accuracy']:.4f}")
print(f"Precision     : {soft_voting_metrics['Precision']:.4f}")
print(f"Recall        : {soft_voting_metrics['Recall']:.4f}")
print(f"F1-Score      : {soft_voting_metrics['F1-Score']:.4f}")
print(f"ROC-AUC       : {soft_voting_metrics['ROC-AUC']:.4f}")


# ============================================================
# E9 — HARD VS SOFT VOTING
# ============================================================

print("\n" + "=" * 90)
print("E9 — HARD VS SOFT VOTING COMPARISON")
print("=" * 90)

voting_comparison_df = pd.DataFrame([
    hard_voting_metrics,
    soft_voting_metrics
])

display(
    voting_comparison_df
    .sort_values("F1-Score", ascending=False)
    .round(4)
)

print("\n📌 Hard Voting uses majority class decisions.")
print("📌 Soft Voting combines predicted class probabilities.")

best_voting_f1_model = voting_comparison_df.loc[
    voting_comparison_df["F1-Score"].idxmax(),
    "Model"
]

best_voting_auc_model = voting_comparison_df.loc[
    voting_comparison_df["ROC-AUC"].idxmax(),
    "Model"
]

print(
    f"\n🏆 Best Voting F1      : {best_voting_f1_model}"
)

print(
    f"🏆 Best Voting ROC-AUC : {best_voting_auc_model}"
)


# ============================================================
# E10 — HARD VS SOFT CONFUSION MATRICES
# ============================================================

print("\n" + "=" * 90)
print("E10 — HARD VS SOFT CONFUSION MATRICES")
print("=" * 90)

fig, axes = plt.subplots(
    1,
    2,
    figsize=(12, 5)
)

ConfusionMatrixDisplay(
    confusion_matrix(
        E_y_cls_test,
        hard_voting_pred
    )
).plot(
    ax=axes[0],
    colorbar=False
)

axes[0].set_title("Hard Voting")

ConfusionMatrixDisplay(
    confusion_matrix(
        E_y_cls_test,
        soft_voting_pred
    )
).plot(
    ax=axes[1],
    colorbar=False
)

axes[1].set_title("Soft Voting")

plt.tight_layout()
plt.show()


# ============================================================
# E11 — VOTING ROC CURVES
# ============================================================

print("\n" + "=" * 90)
print("E11 — VOTING ROC CURVES")
print("=" * 90)

hard_fpr, hard_tpr, _ = roc_curve(
    E_y_cls_test,
    hard_voting_score
)

soft_fpr, soft_tpr, _ = roc_curve(
    E_y_cls_test,
    soft_voting_prob
)

plt.figure(figsize=(10, 7))

plt.plot(
    hard_fpr,
    hard_tpr,
    linewidth=2,
    label=f"Hard Voting (AUC={hard_voting_auc:.4f})"
)

plt.plot(
    soft_fpr,
    soft_tpr,
    linewidth=2,
    label=f"Soft Voting (AUC={soft_voting_metrics['ROC-AUC']:.4f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.5,
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Hard vs Soft Voting — ROC Curves")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# E12 — STACKING CLASSIFIER
# ============================================================

print("\n" + "=" * 90)
print("E12 — STACKING CLASSIFIER")
print("=" * 90)

stacking_classifier = StackingClassifier(
    estimators=E_voting_estimators,
    final_estimator=LogisticRegression(
        max_iter=2000,
        random_state=E_RANDOM_STATE
    ),
    cv=E_CV_FOLDS,
    stack_method="predict_proba",
    n_jobs=-1
)

start = time.perf_counter()

stacking_classifier.fit(
    E_X_cls_train_dense,
    E_y_cls_train
)

stacking_cls_time = time.perf_counter() - start

stacking_cls_pred = stacking_classifier.predict(
    E_X_cls_test_dense
)

stacking_cls_prob = stacking_classifier.predict_proba(
    E_X_cls_test_dense
)[:, 1]

stacking_cls_metrics = {
    "Model": "Stacking Classifier",
    "Accuracy": accuracy_score(
        E_y_cls_test,
        stacking_cls_pred
    ),
    "Precision": precision_score(
        E_y_cls_test,
        stacking_cls_pred,
        zero_division=0
    ),
    "Recall": recall_score(
        E_y_cls_test,
        stacking_cls_pred,
        zero_division=0
    ),
    "F1-Score": f1_score(
        E_y_cls_test,
        stacking_cls_pred,
        zero_division=0
    ),
    "ROC-AUC": roc_auc_score(
        E_y_cls_test,
        stacking_cls_prob
    ),
    "Training Time (sec)": stacking_cls_time
}

print(f"Training Time : {stacking_cls_time:.4f} sec")
print(f"Accuracy      : {stacking_cls_metrics['Accuracy']:.4f}")
print(f"Precision     : {stacking_cls_metrics['Precision']:.4f}")
print(f"Recall        : {stacking_cls_metrics['Recall']:.4f}")
print(f"F1-Score      : {stacking_cls_metrics['F1-Score']:.4f}")
print(f"ROC-AUC       : {stacking_cls_metrics['ROC-AUC']:.4f}")


# ============================================================
# E13 — STACKING CLASSIFIER CONFUSION MATRIX
# ============================================================

print("\n" + "=" * 90)
print("E13 — STACKING CLASSIFIER CONFUSION MATRIX")
print("=" * 90)

ConfusionMatrixDisplay(
    confusion_matrix(
        E_y_cls_test,
        stacking_cls_pred
    )
).plot(
    colorbar=False
)

plt.title("Stacking Classifier — Confusion Matrix")
plt.tight_layout()
plt.show()


# ============================================================
# E14 — STACKING REGRESSOR
# ============================================================

print("\n" + "=" * 90)
print("E14 — STACKING REGRESSOR")
print("=" * 90)

E_stacking_regression_base = [

    (
        "ada",
        AdaBoostRegressor(
            estimator=DecisionTreeRegressor(
                max_depth=3,
                random_state=E_RANDOM_STATE
            ),
            n_estimators=100,
            learning_rate=0.10,
            random_state=E_RANDOM_STATE,
            loss="square"
        )
    ),

    (
        "gb",
        GradientBoostingRegressor(
            n_estimators=100,
            learning_rate=0.10,
            max_depth=3,
            random_state=E_RANDOM_STATE
        )
    ),

    (
        "xgb",
        XGBRegressor(
            n_estimators=100,
            learning_rate=0.10,
            max_depth=4,
            min_child_weight=1,
            subsample=0.90,
            colsample_bytree=0.90,
            reg_alpha=0.10,
            reg_lambda=1.00,
            objective="reg:squarederror",
            eval_metric="rmse",
            random_state=E_RANDOM_STATE,
            n_jobs=-1
        )
    )
]

stacking_regressor = StackingRegressor(
    estimators=E_stacking_regression_base,
    final_estimator=Ridge(
        alpha=1.0
    ),
    cv=E_CV_FOLDS,
    n_jobs=-1
)

start = time.perf_counter()

stacking_regressor.fit(
    E_X_reg_train_dense,
    E_y_reg_train
)

stacking_reg_time = time.perf_counter() - start

stacking_reg_pred = stacking_regressor.predict(
    E_X_reg_test_dense
)

stacking_reg_metrics = {
    "Model": "Stacking Regressor",
    "MAE": mean_absolute_error(
        E_y_reg_test,
        stacking_reg_pred
    ),
    "RMSE": np.sqrt(
        mean_squared_error(
            E_y_reg_test,
            stacking_reg_pred
        )
    ),
    "R²": r2_score(
        E_y_reg_test,
        stacking_reg_pred
    ),
    "Training Time (sec)": stacking_reg_time
}

print(f"Training Time : {stacking_reg_time:.4f} sec")
print(f"MAE           : {stacking_reg_metrics['MAE']:.4f}")
print(f"RMSE          : {stacking_reg_metrics['RMSE']:.4f}")
print(f"R²            : {stacking_reg_metrics['R²']:.4f}")


# ============================================================
# E15 — STACKING REGRESSOR ACTUAL VS PREDICTED
# ============================================================

print("\n" + "=" * 90)
print("E15 — STACKING REGRESSOR ACTUAL VS PREDICTED")
print("=" * 90)

plt.figure(figsize=(9, 7))

plt.scatter(
    E_y_reg_test,
    stacking_reg_pred,
    alpha=0.65
)

min_value = min(
    E_y_reg_test.min(),
    stacking_reg_pred.min()
)

max_value = max(
    E_y_reg_test.max(),
    stacking_reg_pred.max()
)

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--",
    linewidth=2
)

plt.xlabel("Actual Final Score")
plt.ylabel("Predicted Final Score")
plt.title("Stacking Regressor — Actual vs Predicted")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# E16 — LEAKAGE-SAFE STACKING CLASSIFICATION CV
# ============================================================

print("\n" + "=" * 90)
print("E16 — LEAKAGE-SAFE STACKING CLASSIFICATION CV")
print("=" * 90)

if "cv_features" not in globals():
    raise RuntimeError(
        "Part D CV objects were not found. "
        "Run Part D completely before Part E."
    )

if "cv_preprocessor_cls" not in globals():
    raise RuntimeError(
        "cv_preprocessor_cls was not found. "
        "Run the revised Part D completely before Part E."
    )

E_stacking_cv_cls = Pipeline([
    (
        "preprocessor",
        cv_preprocessor_cls
    ),

    (
        "model",
        StackingClassifier(
            estimators=[
                (
                    "ada",
                    AdaBoostClassifier(
                        estimator=DecisionTreeClassifier(
                            max_depth=1,
                            random_state=E_RANDOM_STATE
                        ),
                        n_estimators=100,
                        learning_rate=0.10,
                        random_state=E_RANDOM_STATE
                    )
                ),

                (
                    "gb",
                    GradientBoostingClassifier(
                        n_estimators=100,
                        learning_rate=0.10,
                        max_depth=3,
                        random_state=E_RANDOM_STATE
                    )
                ),

                (
                    "xgb",
                    XGBClassifier(
                        n_estimators=100,
                        learning_rate=0.10,
                        max_depth=4,
                        min_child_weight=1,
                        subsample=0.90,
                        colsample_bytree=0.90,
                        reg_alpha=0.10,
                        reg_lambda=1.00,
                        objective="binary:logistic",
                        eval_metric="logloss",
                        random_state=E_RANDOM_STATE,
                        n_jobs=-1
                    )
                )
            ],

            final_estimator=LogisticRegression(
                max_iter=2000,
                random_state=E_RANDOM_STATE
            ),

            cv=E_CV_FOLDS,
            stack_method="predict_proba",
            n_jobs=-1
        )
    )
])

E_stacking_cls_cv_scores = cross_validate(
    E_stacking_cv_cls,
    cv_features,
    cv_y_cls,

    cv=StratifiedKFold(
        n_splits=E_CV_FOLDS,
        shuffle=True,
        random_state=E_RANDOM_STATE
    ),

    scoring=[
        "accuracy",
        "precision",
        "recall",
        "f1",
        "roc_auc"
    ],

    n_jobs=-1,
    return_train_score=False
)

stacking_classifier_cv_result = pd.DataFrame([{

    "Model": "Stacking Classifier",

    "CV Accuracy":
        E_stacking_cls_cv_scores[
            "test_accuracy"
        ].mean(),

    "CV Precision":
        E_stacking_cls_cv_scores[
            "test_precision"
        ].mean(),

    "CV Recall":
        E_stacking_cls_cv_scores[
            "test_recall"
        ].mean(),

    "CV F1":
        E_stacking_cls_cv_scores[
            "test_f1"
        ].mean(),

    "CV ROC-AUC":
        E_stacking_cls_cv_scores[
            "test_roc_auc"
        ].mean(),

    "CV Accuracy Std":
        E_stacking_cls_cv_scores[
            "test_accuracy"
        ].std(),

    "CV ROC-AUC Std":
        E_stacking_cls_cv_scores[
            "test_roc_auc"
        ].std()
}])

display(
    stacking_classifier_cv_result.round(4)
)

print("✅ Leakage-safe stacking classification CV completed.")


# ============================================================
# E17 — LEAKAGE-SAFE STACKING REGRESSION CV
# ============================================================

print("\n" + "=" * 90)
print("E17 — LEAKAGE-SAFE STACKING REGRESSION CV")
print("=" * 90)

if "cv_preprocessor_reg" not in globals():
    raise RuntimeError(
        "cv_preprocessor_reg was not found. "
        "Run the revised Part D completely before Part E."
    )

E_stacking_cv_reg = Pipeline([
    (
        "preprocessor",
        cv_preprocessor_reg
    ),

    (
        "model",
        StackingRegressor(
            estimators=[
                (
                    "ada",
                    AdaBoostRegressor(
                        estimator=DecisionTreeRegressor(
                            max_depth=3,
                            random_state=E_RANDOM_STATE
                        ),
                        n_estimators=100,
                        learning_rate=0.10,
                        random_state=E_RANDOM_STATE,
                        loss="square"
                    )
                ),

                (
                    "gb",
                    GradientBoostingRegressor(
                        n_estimators=100,
                        learning_rate=0.10,
                        max_depth=3,
                        random_state=E_RANDOM_STATE
                    )
                ),

                (
                    "xgb",
                    XGBRegressor(
                        n_estimators=100,
                        learning_rate=0.10,
                        max_depth=4,
                        min_child_weight=1,
                        subsample=0.90,
                        colsample_bytree=0.90,
                        reg_alpha=0.10,
                        reg_lambda=1.00,
                        objective="reg:squarederror",
                        eval_metric="rmse",
                        random_state=E_RANDOM_STATE,
                        n_jobs=-1
                    )
                )
            ],

            final_estimator=Ridge(
                alpha=1.0
            ),

            cv=E_CV_FOLDS,
            n_jobs=-1
        )
    )
])

E_stacking_reg_cv_scores = cross_validate(
    E_stacking_cv_reg,
    cv_features,
    cv_y_reg,

    cv=KFold(
        n_splits=E_CV_FOLDS,
        shuffle=True,
        random_state=E_RANDOM_STATE
    ),

    scoring=[
        "neg_mean_absolute_error",
        "neg_root_mean_squared_error",
        "r2"
    ],

    n_jobs=-1,
    return_train_score=False
)

stacking_regressor_cv_result = pd.DataFrame([{

    "Model": "Stacking Regressor",

    "CV MAE":
        -E_stacking_reg_cv_scores[
            "test_neg_mean_absolute_error"
        ].mean(),

    "CV RMSE":
        -E_stacking_reg_cv_scores[
            "test_neg_root_mean_squared_error"
        ].mean(),

    "CV R²":
        E_stacking_reg_cv_scores[
            "test_r2"
        ].mean(),

    "CV R² Std":
        E_stacking_reg_cv_scores[
            "test_r2"
        ].std()
}])

display(
    stacking_regressor_cv_result.round(4)
)

print("✅ Leakage-safe stacking regression CV completed.")


# ============================================================
# E18 — CLASSIFICATION ENSEMBLE COMPARISON
# ============================================================

print("\n" + "=" * 90)
print("E18 — COMPLETE CLASSIFICATION ENSEMBLE COMPARISON")
print("=" * 90)

ensemble_classification_results = pd.DataFrame([
    hard_voting_metrics,
    soft_voting_metrics,
    stacking_cls_metrics
])

display(
    ensemble_classification_results
    .sort_values("F1-Score", ascending=False)
    .round(4)
)


# ============================================================
# E19 — REGRESSION ENSEMBLE COMPARISON
# ============================================================

print("\n" + "=" * 90)
print("E19 — COMPLETE REGRESSION ENSEMBLE COMPARISON")
print("=" * 90)

ensemble_regression_results = pd.DataFrame([
    stacking_reg_metrics
])

display(
    ensemble_regression_results.round(4)
)


# ============================================================
# E20 — BEST SINGLE MODEL FROM PART D
# ============================================================

print("\n" + "=" * 90)
print("E20 — BEST SINGLE BOOSTING MODEL")
print("=" * 90)

best_single_cls_idx = boosting_classification_df[
    "F1-Score"
].idxmax()

best_single_cls_name = boosting_classification_df.loc[
    best_single_cls_idx,
    "Model"
]

best_single_cls_row = boosting_classification_df.loc[
    best_single_cls_idx
]

best_single_reg_idx = boosting_regression_df[
    "RMSE"
].idxmin()

best_single_reg_name = boosting_regression_df.loc[
    best_single_reg_idx,
    "Model"
]

best_single_reg_row = boosting_regression_df.loc[
    best_single_reg_idx
]

print(
    f"Best single classification model by F1 : "
    f"{best_single_cls_name}"
)

print(
    f"Best single regression model by RMSE   : "
    f"{best_single_reg_name}"
)


# ============================================================
# E21 — ENSEMBLE IMPROVEMENT
# ============================================================

print("\n" + "=" * 90)
print("E21 — ENSEMBLE IMPROVEMENT ANALYSIS")
print("=" * 90)

best_ensemble_cls_row = ensemble_classification_results.loc[
    ensemble_classification_results["F1-Score"].idxmax()
]

classification_f1_change = (
    best_ensemble_cls_row["F1-Score"]
    -
    best_single_cls_row["F1-Score"]
)

classification_auc_change = (
    best_ensemble_cls_row["ROC-AUC"]
    -
    best_single_cls_row["ROC-AUC"]
)

regression_rmse_change = (
    stacking_reg_metrics["RMSE"]
    -
    best_single_reg_row["RMSE"]
)

regression_r2_change = (
    stacking_reg_metrics["R²"]
    -
    best_single_reg_row["R²"]
)

print(
    f"Classification F1 change  : "
    f"{classification_f1_change:+.4f}"
)

print(
    f"Classification AUC change : "
    f"{classification_auc_change:+.4f}"
)

print(
    f"Regression RMSE change    : "
    f"{regression_rmse_change:+.4f}"
)

print(
    f"Regression R² change      : "
    f"{regression_r2_change:+.4f}"
)


# ============================================================
# E22 — COMPLETE ROC COMPARISON
# ============================================================

print("\n" + "=" * 90)
print("E22 — VOTING + STACKING ROC COMPARISON")
print("=" * 90)

plt.figure(figsize=(10, 7))

# Hard Voting
plt.plot(
    hard_fpr,
    hard_tpr,
    linewidth=2,
    label=f"Hard Voting (AUC={hard_voting_auc:.4f})"
)

# Soft Voting
plt.plot(
    soft_fpr,
    soft_tpr,
    linewidth=2,
    label=f"Soft Voting (AUC={soft_voting_metrics['ROC-AUC']:.4f})"
)

# Stacking
stacking_fpr, stacking_tpr, _ = roc_curve(
    E_y_cls_test,
    stacking_cls_prob
)

plt.plot(
    stacking_fpr,
    stacking_tpr,
    linewidth=2,
    label=f"Stacking (AUC={stacking_cls_metrics['ROC-AUC']:.4f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.5,
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Voting and Stacking — ROC Comparison")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# E23 — TRAINING EFFICIENCY
# ============================================================

print("\n" + "=" * 90)
print("E23 — ENSEMBLE TRAINING EFFICIENCY")
print("=" * 90)

ensemble_training_time_df = pd.DataFrame({

    "Model": [
        "Hard Voting",
        "Soft Voting",
        "Stacking Classifier",
        "Stacking Regressor"
    ],

    "Training Time (sec)": [
        hard_voting_time,
        soft_voting_time,
        stacking_cls_time,
        stacking_reg_time
    ]
})

display(
    ensemble_training_time_df.round(4)
)

plt.figure(figsize=(10, 6))

plt.bar(
    ensemble_training_time_df["Model"],
    ensemble_training_time_df["Training Time (sec)"]
)

plt.ylabel("Training Time (seconds)")
plt.title("Voting and Stacking Training Efficiency")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()


# ============================================================
# E24 — FINAL WINNER ANALYSIS
# ============================================================

print("\n" + "=" * 90)
print("E24 — FINAL ENSEMBLE WINNER ANALYSIS")
print("=" * 90)

best_ensemble_f1_model = ensemble_classification_results.loc[
    ensemble_classification_results["F1-Score"].idxmax(),
    "Model"
]

best_ensemble_auc_model = ensemble_classification_results.loc[
    ensemble_classification_results["ROC-AUC"].idxmax(),
    "Model"
]

best_regression_ensemble_rmse = (
    stacking_reg_metrics["RMSE"]
)

best_regression_ensemble_r2 = (
    stacking_reg_metrics["R²"]
)

print(
    f"🏆 Best Ensemble F1     : "
    f"{best_ensemble_f1_model}"
)

print(
    f"🏆 Best Ensemble ROC-AUC: "
    f"{best_ensemble_auc_model}"
)

print(
    f"🏆 Stacking RMSE        : "
    f"{best_regression_ensemble_rmse:.4f}"
)

print(
    f"🏆 Stacking R²          : "
    f"{best_regression_ensemble_r2:.4f}"
)


# ============================================================
# E25 — FINAL INTERPRETATION
# ============================================================

display(Markdown("""
# 🎯 Part E — Final Conclusion

## Voting

Hard Voting combines final class decisions using majority voting.

Soft Voting combines predicted probabilities and therefore
incorporates model confidence.

The experimental results determine which voting strategy performs
better on this dataset.

## Stacking Classification

The Stacking Classifier uses AdaBoost, Gradient Boosting and XGBoost
as base learners and Logistic Regression as the meta-model.

The meta-model learns how to combine the information generated by
the base learners.

## Stacking Regression

The Stacking Regressor combines AdaBoost, Gradient Boosting and
XGBoost and uses Ridge Regression as the meta-model.

Its performance is evaluated using MAE, RMSE and R².

## Important Ensemble Principle

Greater model complexity does not guarantee better performance.

Therefore, the final selection considers both predictive performance
and practical cost.

An ensemble is preferred only when it provides a meaningful
performance or stability advantage compared with the strongest
individual model.
"""))


# ============================================================
# E26 — DECISION MATRIX
# ============================================================

print("\n" + "=" * 90)
print("E26 — ENSEMBLE DECISION MATRIX")
print("=" * 90)

ensemble_decision_matrix = pd.DataFrame({

    "Criterion": [
        "Best Voting F1",
        "Best Voting ROC-AUC",
        "Best Overall Ensemble F1",
        "Best Overall Ensemble ROC-AUC",
        "Stacking Regression RMSE",
        "Stacking Regression R²"
    ],

    "Best Model": [
        best_voting_f1_model,
        best_voting_auc_model,
        best_ensemble_f1_model,
        best_ensemble_auc_model,
        "Stacking Regressor",
        "Stacking Regressor"
    ]
})

display(
    ensemble_decision_matrix
)


# ============================================================
# E27 — FINAL QUALITY GATE
# ============================================================

print("\n" + "=" * 90)
print("E27 — PART E FINAL QUALITY GATE")
print("=" * 90)

E_final_checks = {

    "Hard Voting Classifier":
        hard_voting is not None,

    "Soft Voting Classifier":
        soft_voting is not None,

    "Stacking Classifier":
        stacking_classifier is not None,

    "Stacking Regressor":
        stacking_regressor is not None,

    "Hard Voting Predictions":
        len(hard_voting_pred) == len(E_y_cls_test),

    "Soft Voting Predictions":
        len(soft_voting_pred) == len(E_y_cls_test),

    "Stacking Classification Predictions":
        len(stacking_cls_pred) == len(E_y_cls_test),

    "Stacking Regression Predictions":
        len(stacking_reg_pred) == len(E_y_reg_test),

    "Hard vs Soft Comparison":
        len(voting_comparison_df) == 2,

    "Classification Ensemble Comparison":
        len(ensemble_classification_results) == 3,

    "Regression Ensemble Comparison":
        len(ensemble_regression_results) == 1,

    "Classification CV":
        len(stacking_classifier_cv_result) == 1,

    "Regression CV":
        len(stacking_regressor_cv_result) == 1,

    "ROC Analysis":
        len(hard_fpr) > 0 and
        len(soft_fpr) > 0 and
        len(stacking_fpr) > 0,

    "Training Efficiency":
        len(ensemble_training_time_df) == 4,

    "Decision Matrix":
        len(ensemble_decision_matrix) == 6
}

all_E_passed = True

for check_name, passed in E_final_checks.items():

    status = (
        "✅ PASSED"
        if passed
        else
        "❌ FAILED"
    )

    print(
        f"{status} | {check_name}"
    )

    if not passed:
        all_E_passed = False

assert all_E_passed, "Part E final quality gate failed."


# ============================================================
# E28 — PRESERVE OBJECTS FOR PART F/G
# ============================================================

ensemble_classification_df = (
    ensemble_classification_results.copy()
)

ensemble_regression_df = (
    ensemble_regression_results.copy()
)

print("\n" + "=" * 90)
print("🎉 PART E — COMPLETED SUCCESSFULLY")
print("=" * 90)

print("""
📦 Models preserved for Part F/G:

• hard_voting
• soft_voting
• stacking_classifier
• stacking_regressor

📊 Result tables preserved:

• voting_comparison_df
• ensemble_classification_df
• ensemble_regression_df
• stacking_classifier_cv_result
• stacking_regressor_cv_result
• ensemble_training_time_df
• ensemble_decision_matrix

🔥 Part E completed successfully!
🚀 Ready for Part F — Evaluation Metrics & Overall Model Comparison!
""")
```

```python
# ============================================================
# 🚀  PART F — EVALUATION METRICS & OVERALL MODEL COMPARISON
# ============================================================
#
# Requirements:
#   • Accuracy
#   • Precision
#   • Recall
#   • F1-Score
#   • ROC-AUC
#   • MAE
#   • RMSE
#   • R²
#   • Overall model comparison
#
# Additional analysis:
#   • Parts C + D + E integration
#   • Classification ranking
#   • Regression ranking
#   • CV comparison
#   • Confusion matrix
#   • ROC comparison
#   • Actual vs predicted
#   • Residual analysis
#   • Generalization
#   • Training efficiency
#   • Final model selection
#   • Quality gate
#
# IMPORTANT:
#   • Uses existing Master Library imports
#   • Does NOT repeat imports
#   • Uses existing models from Parts C, D and E
# ============================================================


# ============================================================
# F1 — OBJECTIVE
# ============================================================

display(Markdown("""
# 📊 Part F — Evaluation Metrics & Overall Model Comparison

## Objective

This section provides a unified evaluation of the machine-learning
models developed throughout the project.

The classification task predicts:

**`completion_status`**

The regression task predicts:

**`final_score`**

Multiple evaluation metrics are used because no single metric can
fully describe model performance.

### Classification Metrics

- Accuracy
- Precision
- Recall
- F1-Score
- ROC-AUC

### Regression Metrics

- Mean Absolute Error (MAE)
- Root Mean Squared Error (RMSE)
- R² Score

The final comparison combines results from:

- Part C — Bagging
- Part D — Boosting
- Part E — Voting and Stacking

The goal is to identify the strongest model for each prediction
task while considering predictive performance, generalization,
stability and computational efficiency.
"""))


# ============================================================
# F2 — METRIC THEORY
# ============================================================

display(Markdown("""
## F2 — Evaluation Metric Theory

### Accuracy

Accuracy measures the proportion of correct predictions among all
predictions.

**Formula:**

Accuracy = Correct Predictions / Total Predictions

Accuracy is useful when class distributions are reasonably balanced,
but it should not be used alone.

---

### Precision

Precision measures how many predicted positive cases were actually
positive.

**Formula:**

Precision = TP / (TP + FP)

Precision is important when false positive predictions are costly.

---

### Recall

Recall measures how many actual positive cases were correctly
identified.

**Formula:**

Recall = TP / (TP + FN)

Recall is especially important when missing a positive case is costly.

---

### F1-Score

F1-Score is the harmonic mean of Precision and Recall.

**Formula:**

F1 = 2 × Precision × Recall / (Precision + Recall)

It provides a balanced measure when both false positives and false
negatives matter.

---

### ROC-AUC

ROC-AUC measures the model's ability to distinguish between the
positive and negative classes across classification thresholds.

A value closer to 1 indicates stronger discrimination.

---

### MAE

Mean Absolute Error measures the average absolute difference between
actual and predicted values.

Lower MAE is better.

---

### RMSE

Root Mean Squared Error gives greater weight to larger prediction
errors.

Lower RMSE is better.

---

### R²

R² measures the proportion of variance in the target explained by
the model.

Higher R² is better.

---

## Important Principle

Different metrics answer different questions.

Therefore, final model selection should consider the complete
evaluation rather than selecting a model using only one score.
"""))


# ============================================================
# F3 — CONFIGURATION
# ============================================================

F_RANDOM_STATE = RANDOM_STATE
F_CV_FOLDS = CV_FOLDS

print("=" * 90)
print("F3 — EVALUATION CONFIGURATION")
print("=" * 90)

print(f"Random State     : {F_RANDOM_STATE}")
print(f"Cross-Validation : {F_CV_FOLDS}-Fold")

print("\nClassification Metrics:")
print("• Accuracy")
print("• Precision")
print("• Recall")
print("• F1-Score")
print("• ROC-AUC")

print("\nRegression Metrics:")
print("• MAE")
print("• RMSE")
print("• R²")


# ============================================================
# F4 — VERIFY REQUIRED MODELS
# ============================================================

print("\n" + "=" * 90)
print("F4 — MODEL AVAILABILITY CHECK")
print("=" * 90)

F_required_objects = [
    "dt_classifier",
    "bag_classifier",
    "ada_classifier",
    "gb_classifier",
    "lgb_classifier",
    "xgb_classifier",
    "hard_voting",
    "soft_voting",
    "stacking_classifier",
    "dt_regressor",
    "bag_regressor",
    "ada_regressor",
    "gb_regressor",
    "lgb_regressor",
    "xgb_regressor",
    "stacking_regressor"
]

F_missing_objects = [
    name
    for name in F_required_objects
    if name not in globals()
]

if F_missing_objects:
    raise RuntimeError(
        "Missing required objects: "
        + ", ".join(F_missing_objects)
    )

for name in F_required_objects:
    print(f"✅ Available | {name}")

print("\n✅ All Part C, D and E models are available.")


# ============================================================
# F5 — CLASSIFICATION PREDICTIONS
# CORRECTED VERSION — HARD VOTING SAFE
# ============================================================

print("\n" + "=" * 90)
print("F5 — CLASSIFICATION PREDICTIONS")
print("=" * 90)

F_classification_predictions = {}
F_classification_probabilities = {}

for model_name, model in F_classification_models.items():

    # Actual predicted classes
    F_classification_predictions[model_name] = model.predict(
        E_X_cls_test_dense
    )

    # Hard Voting does NOT expose predict_proba().
    # Therefore, calculate a continuous probability score by
    # averaging the positive-class probabilities of its fitted
    # base estimators.
    if model_name == "Hard Voting":

        base_probabilities = np.column_stack([
            estimator.predict_proba(
                E_X_cls_test_dense
            )[:, 1]
            for estimator in model.estimators_
        ])

        F_classification_probabilities[model_name] = (
            base_probabilities.mean(axis=1)
        )

    else:

        F_classification_probabilities[model_name] = (
            model.predict_proba(
                E_X_cls_test_dense
            )[:, 1]
        )

    print(
        f"✅ Predictions generated | {model_name}"
    )

print("\n✅ Classification prediction generation completed.")
print("✅ Hard Voting handled without calling predict_proba().")


# ============================================================
# F6 — CLASSIFICATION METRICS
# ============================================================

print("\n" + "=" * 90)
print("F6 — CLASSIFICATION METRICS")
print("=" * 90)

F_classification_rows = []

for model_name in F_classification_models:

    predictions = F_classification_predictions[
        model_name
    ]

    probabilities = F_classification_probabilities[
        model_name
    ]

    F_classification_rows.append({

        "Model": model_name,

        "Accuracy": accuracy_score(
            E_y_cls_test,
            predictions
        ),

        "Precision": precision_score(
            E_y_cls_test,
            predictions,
            zero_division=0
        ),

        "Recall": recall_score(
            E_y_cls_test,
            predictions,
            zero_division=0
        ),

        "F1-Score": f1_score(
            E_y_cls_test,
            predictions,
            zero_division=0
        ),

        "ROC-AUC": roc_auc_score(
            E_y_cls_test,
            probabilities
        )
    })


classification_metrics_df = pd.DataFrame(
    F_classification_rows
)

display(
    classification_metrics_df
    .sort_values(
        "F1-Score",
        ascending=False
    )
    .round(4)
)


# ============================================================
# F7 — CLASSIFICATION METRIC WINNERS
# ============================================================

print("\n" + "=" * 90)
print("F7 — CLASSIFICATION METRIC WINNERS")
print("=" * 90)

F_classification_metric_directions = {
    "Accuracy": "max",
    "Precision": "max",
    "Recall": "max",
    "F1-Score": "max",
    "ROC-AUC": "max"
}

classification_metric_winners = []

for metric, direction in F_classification_metric_directions.items():

    if direction == "max":
        idx = classification_metrics_df[
            metric
        ].idxmax()

    winner = classification_metrics_df.loc[
        idx,
        "Model"
    ]

    value = classification_metrics_df.loc[
        idx,
        metric
    ]

    classification_metric_winners.append({

        "Metric": metric,

        "Best Model": winner,

        "Score": value
    })

    print(
        f"🏆 {metric:<12} : "
        f"{winner:<22} "
        f"({value:.4f})"
    )


classification_metric_winners_df = pd.DataFrame(
    classification_metric_winners
)


# ============================================================
# F8 — CLASSIFICATION PERFORMANCE RANKING
# ============================================================

print("\n" + "=" * 90)
print("F8 — CLASSIFICATION PERFORMANCE RANKING")
print("=" * 90)

F_cls_rank = classification_metrics_df.copy()

F_cls_rank["Mean Core Score"] = F_cls_rank[
    [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score",
        "ROC-AUC"
    ]
].mean(axis=1)

classification_ranking_df = (
    F_cls_rank
    .sort_values(
        "Mean Core Score",
        ascending=False
    )
    .reset_index(drop=True)
)

classification_ranking_df.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(classification_ranking_df) + 1
    )
)

display(
    classification_ranking_df.round(4)
)


# ============================================================
# F9 — CLASSIFICATION CONFUSION MATRIX
# ============================================================

print("\n" + "=" * 90)
print("F9 — BEST CLASSIFICATION MODEL CONFUSION MATRIX")
print("=" * 90)

F_best_cls_model = classification_metrics_df.loc[
    classification_metrics_df["F1-Score"].idxmax(),
    "Model"
]

F_best_cls_pred = F_classification_predictions[
    F_best_cls_model
]

print(
    f"Best classification model by F1: "
    f"{F_best_cls_model}"
)

ConfusionMatrixDisplay(
    confusion_matrix(
        E_y_cls_test,
        F_best_cls_pred
    )
).plot(
    colorbar=False
)

plt.title(
    f"{F_best_cls_model} — Confusion Matrix"
)

plt.tight_layout()
plt.show()


# ============================================================
# F10 — COMPLETE ROC COMPARISON
# ============================================================

print("\n" + "=" * 90)
print("F10 — COMPLETE CLASSIFICATION ROC COMPARISON")
print("=" * 90)

plt.figure(figsize=(11, 8))

for model_name in F_classification_models:

    probabilities = F_classification_probabilities[
        model_name
    ]

    fpr, tpr, _ = roc_curve(
        E_y_cls_test,
        probabilities
    )

    auc_value = roc_auc_score(
        E_y_cls_test,
        probabilities
    )

    plt.plot(
        fpr,
        tpr,
        linewidth=2,
        label=f"{model_name} (AUC={auc_value:.4f})"
    )


plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.5,
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC-AUC Comparison — All Classification Models")
plt.legend(
    bbox_to_anchor=(1.04, 1),
    loc="upper left"
)

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# F11 — REGRESSION PREDICTIONS
# ============================================================

print("\n" + "=" * 90)
print("F11 — REGRESSION PREDICTIONS")
print("=" * 90)

F_regression_models = {

    "Decision Tree": dt_regressor,

    "Bagging Regressor": bag_regressor,

    "AdaBoost": ada_regressor,

    "Gradient Boosting": gb_regressor,

    "LightGBM": lgb_regressor,

    "XGBoost": xgb_regressor,

    "Stacking Regressor": stacking_regressor
}


F_regression_predictions = {}

for model_name, model in F_regression_models.items():

    F_regression_predictions[model_name] = (
        model.predict(
            E_X_reg_test_dense
        )
    )

    print(
        f"✅ Predictions generated | {model_name}"
    )


# ============================================================
# F12 — REGRESSION METRICS
# ============================================================

print("\n" + "=" * 90)
print("F12 — REGRESSION METRICS")
print("=" * 90)

F_regression_rows = []

for model_name in F_regression_models:

    predictions = F_regression_predictions[
        model_name
    ]

    F_regression_rows.append({

        "Model": model_name,

        "MAE": mean_absolute_error(
            E_y_reg_test,
            predictions
        ),

        "RMSE": np.sqrt(
            mean_squared_error(
                E_y_reg_test,
                predictions
            )
        ),

        "R²": r2_score(
            E_y_reg_test,
            predictions
        )
    })


regression_metrics_df = pd.DataFrame(
    F_regression_rows
)

display(
    regression_metrics_df
    .sort_values(
        "RMSE",
        ascending=True
    )
    .round(4)
)


# ============================================================
# F13 — REGRESSION METRIC WINNERS
# ============================================================

print("\n" + "=" * 90)
print("F13 — REGRESSION METRIC WINNERS")
print("=" * 90)

F_regression_metric_directions = {
    "MAE": "min",
    "RMSE": "min",
    "R²": "max"
}

regression_metric_winners = []

for metric, direction in F_regression_metric_directions.items():

    if direction == "min":
        idx = regression_metrics_df[
            metric
        ].idxmin()

    else:
        idx = regression_metrics_df[
            metric
        ].idxmax()

    winner = regression_metrics_df.loc[
        idx,
        "Model"
    ]

    value = regression_metrics_df.loc[
        idx,
        metric
    ]

    regression_metric_winners.append({

        "Metric": metric,

        "Best Model": winner,

        "Score": value
    })

    print(
        f"🏆 {metric:<6} : "
        f"{winner:<22} "
        f"({value:.4f})"
    )


regression_metric_winners_df = pd.DataFrame(
    regression_metric_winners
)


# ============================================================
# F14 — REGRESSION PERFORMANCE RANKING
# ============================================================

print("\n" + "=" * 90)
print("F14 — REGRESSION PERFORMANCE RANKING")
print("=" * 90)

F_reg_rank = regression_metrics_df.copy()

# Normalize each metric so that higher normalized score
# always means better performance.

def minmax_score(series, higher_is_better=True):

    minimum = series.min()
    maximum = series.max()

    if maximum == minimum:
        return pd.Series(
            np.ones(len(series)),
            index=series.index
        )

    normalized = (
        (series - minimum)
        /
        (maximum - minimum)
    )

    if higher_is_better:
        return normalized

    return 1 - normalized


F_reg_rank["MAE Score"] = minmax_score(
    F_reg_rank["MAE"],
    higher_is_better=False
)

F_reg_rank["RMSE Score"] = minmax_score(
    F_reg_rank["RMSE"],
    higher_is_better=False
)

F_reg_rank["R² Score"] = minmax_score(
    F_reg_rank["R²"],
    higher_is_better=True
)

F_reg_rank["Mean Core Score"] = F_reg_rank[
    [
        "MAE Score",
        "RMSE Score",
        "R² Score"
    ]
].mean(axis=1)

regression_ranking_df = (
    F_reg_rank
    .sort_values(
        "Mean Core Score",
        ascending=False
    )
    .reset_index(drop=True)
)

regression_ranking_df.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(regression_ranking_df) + 1
    )
)

display(
    regression_ranking_df.round(4)
)


# ============================================================
# F15 — BEST REGRESSION ACTUAL VS PREDICTED
# ============================================================

print("\n" + "=" * 90)
print("F15 — BEST REGRESSION MODEL")
print("=" * 90)

F_best_reg_model = regression_metrics_df.loc[
    regression_metrics_df["RMSE"].idxmin(),
    "Model"
]

F_best_reg_pred = F_regression_predictions[
    F_best_reg_model
]

print(
    f"Best regression model by RMSE: "
    f"{F_best_reg_model}"
)

plt.figure(figsize=(9, 7))

plt.scatter(
    E_y_reg_test,
    F_best_reg_pred,
    alpha=0.65
)

F_min_value = min(
    E_y_reg_test.min(),
    F_best_reg_pred.min()
)

F_max_value = max(
    E_y_reg_test.max(),
    F_best_reg_pred.max()
)

plt.plot(
    [F_min_value, F_max_value],
    [F_min_value, F_max_value],
    linestyle="--",
    linewidth=2
)

plt.xlabel("Actual Final Score")
plt.ylabel("Predicted Final Score")

plt.title(
    f"{F_best_reg_model} — Actual vs Predicted"
)

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# F16 — RESIDUAL ANALYSIS
# ============================================================

print("\n" + "=" * 90)
print("F16 — REGRESSION RESIDUAL ANALYSIS")
print("=" * 90)

F_residuals = (
    E_y_reg_test.values
    -
    F_best_reg_pred
)

print(
    f"Mean Residual : {F_residuals.mean():.4f}"
)

print(
    f"Residual Std  : {F_residuals.std():.4f}"
)

plt.figure(figsize=(10, 6))

plt.scatter(
    F_best_reg_pred,
    F_residuals,
    alpha=0.65
)

plt.axhline(
    0,
    linestyle="--",
    linewidth=2
)

plt.xlabel("Predicted Final Score")
plt.ylabel("Residual (Actual − Predicted)")

plt.title(
    f"{F_best_reg_model} — Residual Analysis"
)

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# F17 — CROSS-VALIDATION SUMMARY
# ============================================================

print("\n" + "=" * 90)
print("F17 — CROSS-VALIDATION SUMMARY")
print("=" * 90)

display(Markdown("""
## Cross-Validation Interpretation

Cross-validation results are used alongside the held-out test-set
results.

A strong model should ideally demonstrate:

- Strong mean performance
- Low variation between folds
- Similar behaviour on unseen test data

Cross-validation therefore provides additional evidence about
generalization and stability.
"""))


# Collect available CV results from Parts C, D and E.

F_cv_classification_tables = []

if "classification_cv_df" in globals():

    F_temp = classification_cv_df.copy()

    F_temp["Source"] = "Part C"

    F_cv_classification_tables.append(
        F_temp
    )


if "boosting_cls_cv_df" in globals():

    F_temp = boosting_cls_cv_df.copy()

    F_temp["Source"] = "Part D"

    F_cv_classification_tables.append(
        F_temp
    )


if "stacking_classifier_cv_result" in globals():

    F_temp = stacking_classifier_cv_result.copy()

    F_temp["Source"] = "Part E"

    F_cv_classification_tables.append(
        F_temp
    )


if F_cv_classification_tables:

    F_combined_cls_cv = pd.concat(
        F_cv_classification_tables,
        ignore_index=True
    )

    print(
        "Classification CV results from Parts C, D and E:"
    )

    display(
        F_combined_cls_cv.round(4)
    )


# ============================================================
# F18 — REGRESSION CV SUMMARY
# ============================================================

F_cv_regression_tables = []

if "regression_cv_df" in globals():

    F_temp = regression_cv_df.copy()

    F_temp["Source"] = "Part C"

    F_cv_regression_tables.append(
        F_temp
    )


if "boosting_reg_cv_df" in globals():

    F_temp = boosting_reg_cv_df.copy()

    F_temp["Source"] = "Part D"

    F_cv_regression_tables.append(
        F_temp
    )


if "stacking_regressor_cv_result" in globals():

    F_temp = stacking_regressor_cv_result.copy()

    F_temp["Source"] = "Part E"

    F_cv_regression_tables.append(
        F_temp
    )


if F_cv_regression_tables:

    F_combined_reg_cv = pd.concat(
        F_cv_regression_tables,
        ignore_index=True
    )

    print(
        "Regression CV results from Parts C, D and E:"
    )

    display(
        F_combined_reg_cv.round(4)
    )


# ============================================================
# F19 — TEST-SET GENERALIZATION GAP
# ============================================================

print("\n" + "=" * 90)
print("F19 — GENERALIZATION ANALYSIS")
print("=" * 90)

F_generalization_rows = []

# Classification generalization based on train/test accuracy
# where available from trained models.

for model_name, model in F_classification_models.items():

    train_pred = model.predict(
        E_X_cls_train_dense
    )

    test_pred = F_classification_predictions[
        model_name
    ]

    train_accuracy = accuracy_score(
        E_y_cls_train,
        train_pred
    )

    test_accuracy = accuracy_score(
        E_y_cls_test,
        test_pred
    )

    F_generalization_rows.append({

        "Task": "Classification",

        "Model": model_name,

        "Train Accuracy": train_accuracy,

        "Test Accuracy": test_accuracy,

        "Accuracy Gap":
            train_accuracy - test_accuracy
    })


classification_generalization_f_df = pd.DataFrame(
    F_generalization_rows
)

display(
    classification_generalization_f_df
    .sort_values(
        "Accuracy Gap",
        ascending=True
    )
    .round(4)
)


# ============================================================
# F20 — OVERALL MODEL SCORECARD
# ============================================================

print("\n" + "=" * 90)
print("F20 — OVERALL MODEL SCORECARD")
print("=" * 90)

display(Markdown("""
# 🏆 Overall Model Scorecard

The final scorecard combines the principal evaluation dimensions.

### Classification

Primary considerations:

1. F1-Score
2. ROC-AUC
3. Accuracy
4. Precision
5. Recall

### Regression

Primary considerations:

1. RMSE
2. MAE
3. R²

The ranking is used as a decision-support tool rather than as an
absolute universal measure.
"""))


# ============================================================
# F21 — FINAL CLASSIFICATION DECISION
# ============================================================

print("\n" + "=" * 90)
print("F21 — FINAL CLASSIFICATION DECISION")
print("=" * 90)

F_cls_f1_winner = classification_metrics_df.loc[
    classification_metrics_df["F1-Score"].idxmax()
]

F_cls_auc_winner = classification_metrics_df.loc[
    classification_metrics_df["ROC-AUC"].idxmax()
]

F_cls_accuracy_winner = classification_metrics_df.loc[
    classification_metrics_df["Accuracy"].idxmax()
]

print(
    f"🥇 Best F1       : "
    f"{F_cls_f1_winner['Model']} "
    f"({F_cls_f1_winner['F1-Score']:.4f})"
)

print(
    f"🥇 Best ROC-AUC   : "
    f"{F_cls_auc_winner['Model']} "
    f"({F_cls_auc_winner['ROC-AUC']:.4f})"
)

print(
    f"🥇 Best Accuracy  : "
    f"{F_cls_accuracy_winner['Model']} "
    f"({F_cls_accuracy_winner['Accuracy']:.4f})"
)


# ============================================================
# F22 — FINAL REGRESSION DECISION
# ============================================================

print("\n" + "=" * 90)
print("F22 — FINAL REGRESSION DECISION")
print("=" * 90)

F_reg_rmse_winner = regression_metrics_df.loc[
    regression_metrics_df["RMSE"].idxmin()
]

F_reg_mae_winner = regression_metrics_df.loc[
    regression_metrics_df["MAE"].idxmin()
]

F_reg_r2_winner = regression_metrics_df.loc[
    regression_metrics_df["R²"].idxmax()
]

print(
    f"🥇 Best RMSE : "
    f"{F_reg_rmse_winner['Model']} "
    f"({F_reg_rmse_winner['RMSE']:.4f})"
)

print(
    f"🥇 Best MAE  : "
    f"{F_reg_mae_winner['Model']} "
    f"({F_reg_mae_winner['MAE']:.4f})"
)

print(
    f"🥇 Best R²   : "
    f"{F_reg_r2_winner['Model']} "
    f"({F_reg_r2_winner['R²']:.4f})"
)


# ============================================================
# F23 — FINAL INTERPRETATION
# ============================================================

display(Markdown("""
# 🎯 Part F — Final Evaluation Interpretation

## Classification

Classification models were compared using Accuracy, Precision,
Recall, F1-Score and ROC-AUC.

Because the project involves predicting whether a student completes
a course, F1-Score and ROC-AUC are particularly useful alongside
Accuracy.

The final classification model should therefore be selected based
on the balance between predictive performance, generalization and
practical requirements.

## Regression

Regression models were compared using MAE, RMSE and R².

RMSE is useful for highlighting larger prediction errors, while MAE
provides a direct interpretation of average absolute prediction
error.

R² provides an overall measure of explained variance.

The final regression model should therefore be selected using all
three metrics rather than relying on only one.

## Overall Learning

The experiments across Parts C, D and E demonstrate that:

- Bagging can reduce variance through bootstrap aggregation.
- Boosting can sequentially improve weak learners.
- Voting can combine independent model decisions.
- Stacking can learn how to combine base-model predictions.
- More complex models do not automatically provide better results.
- Model selection should consider performance, stability,
  generalization and computational cost.

The final model choice is therefore evidence-based rather than
based solely on model complexity or a single metric.
"""))


# ============================================================
# F24 — FINAL DECISION MATRIX
# ============================================================

print("\n" + "=" * 90)
print("F24 — FINAL DECISION MATRIX")
print("=" * 90)

F_final_decision_matrix = pd.DataFrame({

    "Evaluation Criterion": [

        "Best Classification Accuracy",

        "Best Classification Precision",

        "Best Classification Recall",

        "Best Classification F1",

        "Best Classification ROC-AUC",

        "Best Regression MAE",

        "Best Regression RMSE",

        "Best Regression R²"
    ],

    "Winning Model": [

        classification_metrics_df.loc[
            classification_metrics_df[
                "Accuracy"
            ].idxmax(),
            "Model"
        ],

        classification_metrics_df.loc[
            classification_metrics_df[
                "Precision"
            ].idxmax(),
            "Model"
        ],

        classification_metrics_df.loc[
            classification_metrics_df[
                "Recall"
            ].idxmax(),
            "Model"
        ],

        classification_metrics_df.loc[
            classification_metrics_df[
                "F1-Score"
            ].idxmax(),
            "Model"
        ],

        classification_metrics_df.loc[
            classification_metrics_df[
                "ROC-AUC"
            ].idxmax(),
            "Model"
        ],

        regression_metrics_df.loc[
            regression_metrics_df[
                "MAE"
            ].idxmin(),
            "Model"
        ],

        regression_metrics_df.loc[
            regression_metrics_df[
                "RMSE"
            ].idxmin(),
            "Model"
        ],

        regression_metrics_df.loc[
            regression_metrics_df[
                "R²"
            ].idxmax(),
            "Model"
        ]
    ]
})

display(
    F_final_decision_matrix
)


# ============================================================
# F25 — PRESERVE FINAL RESULTS FOR PART G
# ============================================================

F_final_classification_results = (
    classification_metrics_df.copy()
)

F_final_regression_results = (
    regression_metrics_df.copy()
)

F_final_classification_ranking = (
    classification_ranking_df.copy()
)

F_final_regression_ranking = (
    regression_ranking_df.copy()
)

F_final_decision_matrix = (
    F_final_decision_matrix.copy()
)


# ============================================================
# F26 — FINAL QUALITY GATE
# ============================================================

print("\n" + "=" * 90)
print("F26 — PART F FINAL QUALITY GATE")
print("=" * 90)

F_quality_checks = {

    "Classification metrics":
        len(classification_metrics_df) == 9,

    "Regression metrics":
        len(regression_metrics_df) == 7,

    "Accuracy calculated":
        classification_metrics_df["Accuracy"].notna().all(),

    "Precision calculated":
        classification_metrics_df["Precision"].notna().all(),

    "Recall calculated":
        classification_metrics_df["Recall"].notna().all(),

    "F1 calculated":
        classification_metrics_df["F1-Score"].notna().all(),

    "ROC-AUC calculated":
        classification_metrics_df["ROC-AUC"].notna().all(),

    "MAE calculated":
        regression_metrics_df["MAE"].notna().all(),

    "RMSE calculated":
        regression_metrics_df["RMSE"].notna().all(),

    "R² calculated":
        regression_metrics_df["R²"].notna().all(),

    "Classification ranking":
        len(classification_ranking_df) == 9,

    "Regression ranking":
        len(regression_ranking_df) == 7,

    "Classification decision matrix":
        len(F_final_decision_matrix) == 8,

    "Classification predictions":
        all(
            len(pred) == len(E_y_cls_test)
            for pred in F_classification_predictions.values()
        ),

    "Regression predictions":
        all(
            len(pred) == len(E_y_reg_test)
            for pred in F_regression_predictions.values()
        )
}

F_all_passed = True

for check_name, passed in F_quality_checks.items():

    print(
        f"{'✅ PASSED' if passed else '❌ FAILED'} | "
        f"{check_name}"
    )

    if not passed:
        F_all_passed = False


assert F_all_passed, (
    "Part F final quality gate failed."
)


# ============================================================
# F27 — COMPLETION
# ============================================================

print("\n" + "=" * 90)
print("🎉 PART F — COMPLETED SUCCESSFULLY")
print("=" * 90)

print("""
📊 Final evaluation tables preserved for Part G:

• F_final_classification_results
• F_final_regression_results
• F_final_classification_ranking
• F_final_regression_ranking
• F_final_decision_matrix

📈 Analysis completed:

• Classification metrics
• Regression metrics
• Metric winners
• Model rankings
• Confusion matrix
• ROC comparison
• Actual vs predicted
• Residual analysis
• Cross-validation summary
• Generalization analysis
• Final decision matrix

🔥 Part F completed successfully!
🚀 Ready for Part G — Final Report, Business Interpretation & Conclusions!
""")
```

```python
# ============================================================
# 🚀 PART G — FINAL REPORT, BUSINESS INTERPRETATION & CONCLUSIONS
# ============================================================
#
# Purpose:
#   • Consolidate Parts B–F
#   • Present final model performance
#   • Interpret results
#   • Provide business/academic implications
#   • Discuss limitations
#   • Provide final recommendations
#   • Generate final tables and visualizations
#   • Preserve all important outputs
#
# IMPORTANT:
#   • Uses existing Master Library imports
#   • Does NOT repeat imports
#   • Uses existing results from Parts C–F
#   • No new models are introduced
# ============================================================


# ============================================================
# G1 — EXECUTIVE SUMMARY
# ============================================================

display(Markdown("""
# 🏆 Part G — Final Report, Business Interpretation & Conclusions

## Smart Outcome Predictor

### Executive Summary

The **Smart Outcome Predictor** project applies ensemble learning
techniques to predict two student-learning outcomes:

1. **Course Completion Status** — classification task
2. **Final Score** — regression task

The project systematically evaluated multiple ensemble approaches,
including Bagging, AdaBoost, Gradient Boosting, LightGBM, XGBoost,
Voting and Stacking.

The experiments demonstrate that ensemble learning generally
provides stronger predictive performance than a single Decision
Tree model.

For classification, **Hard Voting** achieved the strongest overall
held-out test-set performance based on Accuracy, F1-Score and
ROC-AUC.

For regression, **Stacking Regressor** achieved the lowest RMSE and
highest R², while **Gradient Boosting** achieved the lowest MAE.
However, the numerical difference between these two models is
small, so the result should be interpreted as a marginal advantage
rather than a dramatic improvement.

The final analysis also demonstrates the importance of evaluating
generalization, stability and computational cost rather than
selecting a model solely because it has high training performance.
"""))


# ============================================================
# G2 — PROJECT OBJECTIVES
# ============================================================

display(Markdown("""
## G2 — Project Objectives

The main objectives of the Smart Outcome Predictor are:

- Understand and prepare a real-world educational dataset.
- Predict whether a student completes a course.
- Predict a student's final course score.
- Apply Bagging and Boosting ensemble techniques.
- Implement LightGBM and XGBoost.
- Compare Voting and Stacking ensembles.
- Evaluate models using appropriate classification and regression
  metrics.
- Analyse model generalization and stability.
- Identify important predictive factors.
- Select practical final models based on experimental evidence.
"""))


# ============================================================
# G3 — DATASET SUMMARY
# ============================================================

display(Markdown("""
## G3 — Dataset Summary

The dataset contains **5,200 student records** and **19 original
columns**.

### Classification Target

`completion_status`

This target represents whether a student completed the course.

### Regression Target

`final_score`

This target represents the student's final course score.

### Important Predictor Categories

The dataset contains information related to:

- Student demographics
- Country/region
- Device type
- Educational background
- Course level
- Course category
- Course timing
- Learning sessions
- Time spent
- Videos watched
- Quiz attempts
- Assignment submissions
- Forum participation
- Average quiz score
- Attendance rate

An ID-like identifier was excluded from model training to avoid
using non-predictive identification information.
"""))


# ============================================================
# G4 — DATA PREPARATION SUMMARY
# ============================================================

display(Markdown("""
## G4 — Data Preparation & Methodology

The modelling workflow included:

1. Dataset loading
2. Data-quality inspection
3. Missing-value analysis
4. Date feature engineering
5. Removal of ID-like and target columns from predictors
6. Train-test splitting
7. Numerical imputation
8. Numerical scaling
9. Categorical imputation
10. One-hot encoding
11. Model training
12. Test-set evaluation
13. Cross-validation
14. Model comparison

The preprocessing pipeline was fitted using the training data and
then applied to the test data. This helps prevent information from
the test set from influencing model training.

The processed feature representation contained **34 features**.
"""))


# ============================================================
# G5 — EXPERIMENTAL COVERAGE
# ============================================================

print("\n" + "=" * 90)
print("G5 — EXPERIMENTAL COVERAGE")
print("=" * 90)

G_classification_models = [
    "Decision Tree",
    "Bagging Classifier",
    "AdaBoost",
    "Gradient Boosting",
    "LightGBM",
    "XGBoost",
    "Hard Voting",
    "Soft Voting",
    "Stacking Classifier"
]

G_regression_models = [
    "Decision Tree",
    "Bagging Regressor",
    "AdaBoost",
    "Gradient Boosting",
    "LightGBM",
    "XGBoost",
    "Stacking Regressor"
]

print("Classification Models:")
for i, model in enumerate(G_classification_models, 1):
    print(f"{i:>2}. {model}")

print("\nRegression Models:")
for i, model in enumerate(G_regression_models, 1):
    print(f"{i:>2}. {model}")

print(
    "\n✅ Comprehensive ensemble-learning comparison completed."
)


# ============================================================
# G6 — FINAL CLASSIFICATION TABLE
# ============================================================

print("\n" + "=" * 90)
print("G6 — FINAL CLASSIFICATION RESULTS")
print("=" * 90)

G_classification_final = (
    F_final_classification_results
    .copy()
    .sort_values(
        "F1-Score",
        ascending=False
    )
    .reset_index(drop=True)
)

G_classification_final.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(G_classification_final) + 1
    )
)

display(
    G_classification_final.round(4)
)


# ============================================================
# G7 — FINAL REGRESSION TABLE
# ============================================================

print("\n" + "=" * 90)
print("G7 — FINAL REGRESSION RESULTS")
print("=" * 90)

G_regression_final = (
    F_final_regression_results
    .copy()
    .sort_values(
        "RMSE",
        ascending=True
    )
    .reset_index(drop=True)
)

G_regression_final.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(G_regression_final) + 1
    )
)

display(
    G_regression_final.round(4)
)


# ============================================================
# G8 — CLASSIFICATION VISUAL COMPARISON
# ============================================================

print("\n" + "=" * 90)
print("G8 — CLASSIFICATION PERFORMANCE VISUALIZATION")
print("=" * 90)

G_cls_plot = G_classification_final.copy()

plt.figure(figsize=(12, 7))

plt.bar(
    G_cls_plot["Model"],
    G_cls_plot["F1-Score"]
)

plt.xlabel("Model")
plt.ylabel("F1-Score")
plt.title("Final Classification Model Comparison — F1-Score")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()
plt.show()


# ============================================================
# G9 — ROC-AUC VISUAL COMPARISON
# ============================================================

print("\n" + "=" * 90)
print("G9 — ROC-AUC COMPARISON")
print("=" * 90)

G_auc_plot = (
    G_classification_final
    .sort_values(
        "ROC-AUC",
        ascending=False
    )
)

plt.figure(figsize=(12, 7))

plt.bar(
    G_auc_plot["Model"],
    G_auc_plot["ROC-AUC"]
)

plt.xlabel("Model")
plt.ylabel("ROC-AUC")
plt.title("Final Classification Model Comparison — ROC-AUC")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()
plt.show()


# ============================================================
# G10 — REGRESSION VISUAL COMPARISON
# ============================================================

print("\n" + "=" * 90)
print("G10 — REGRESSION RMSE COMPARISON")
print("=" * 90)

G_reg_plot = (
    G_regression_final
    .sort_values(
        "RMSE",
        ascending=True
    )
)

plt.figure(figsize=(11, 7))

plt.bar(
    G_reg_plot["Model"],
    G_reg_plot["RMSE"]
)

plt.xlabel("Model")
plt.ylabel("RMSE")
plt.title("Final Regression Model Comparison — RMSE")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()
plt.show()


# ============================================================
# G11 — FINAL WINNERS
# ============================================================

print("\n" + "=" * 90)
print("G11 — FINAL MODEL WINNERS")
print("=" * 90)

G_best_cls_f1 = G_classification_final.loc[
    G_classification_final["F1-Score"].idxmax()
]

G_best_cls_auc = G_classification_final.loc[
    G_classification_final["ROC-AUC"].idxmax()
]

G_best_cls_accuracy = G_classification_final.loc[
    G_classification_final["Accuracy"].idxmax()
]

G_best_reg_mae = G_regression_final.loc[
    G_regression_final["MAE"].idxmin()
]

G_best_reg_rmse = G_regression_final.loc[
    G_regression_final["RMSE"].idxmin()
]

G_best_reg_r2 = G_regression_final.loc[
    G_regression_final["R²"].idxmax()
]

print(
    f"🏆 Classification — Accuracy : "
    f"{G_best_cls_accuracy['Model']} "
    f"({G_best_cls_accuracy['Accuracy']:.4f})"
)

print(
    f"🏆 Classification — F1       : "
    f"{G_best_cls_f1['Model']} "
    f"({G_best_cls_f1['F1-Score']:.4f})"
)

print(
    f"🏆 Classification — ROC-AUC  : "
    f"{G_best_cls_auc['Model']} "
    f"({G_best_cls_auc['ROC-AUC']:.4f})"
)

print(
    f"🏆 Regression — MAE          : "
    f"{G_best_reg_mae['Model']} "
    f"({G_best_reg_mae['MAE']:.4f})"
)

print(
    f"🏆 Regression — RMSE         : "
    f"{G_best_reg_rmse['Model']} "
    f"({G_best_reg_rmse['RMSE']:.4f})"
)

print(
    f"🏆 Regression — R²           : "
    f"{G_best_reg_r2['Model']} "
    f"({G_best_reg_r2['R²']:.4f})"
)


# ============================================================
# G12 — CLASSIFICATION INTERPRETATION
# ============================================================

display(Markdown("""
## G12 — Classification Results Interpretation

### Final Classification Model: Hard Voting

Hard Voting produced the strongest overall classification result
in the held-out test-set comparison.

Its performance was:

- **Accuracy:** 0.7442
- **Precision:** 0.6856
- **Recall:** 0.5872
- **F1-Score:** 0.6326
- **ROC-AUC:** 0.7909

Hard Voting therefore provides the best combined performance in
the current experiment when Accuracy, F1-Score and ROC-AUC are
considered together.

However, the advantage over XGBoost is modest. XGBoost achieved an
F1-Score of 0.6287 and ROC-AUC of 0.7874.

Therefore, the result should be described as a **modest improvement**
rather than a dramatic performance gain.

### Other Classification Findings

- Stacking Classifier achieved the highest Precision: **0.6909**.
- XGBoost achieved the highest Recall: **0.5949**.
- Gradient Boosting achieved very similar ROC-AUC performance:
  **0.7903**.

This demonstrates that model selection depends on the evaluation
objective and that different models can be preferable for different
business requirements.
"""))


# ============================================================
# G13 — REGRESSION INTERPRETATION
# ============================================================

display(Markdown("""
## G13 — Regression Results Interpretation

### Final Regression Model: Stacking Regressor

Stacking Regressor achieved:

- **MAE:** 7.8793
- **RMSE:** 9.8007
- **R²:** 0.4859

It achieved the best RMSE and R² among the evaluated regression
models.

However, Gradient Boosting achieved the lowest MAE:

- **MAE:** 7.8668
- **RMSE:** 9.8176
- **R²:** 0.4841

The difference between Stacking Regressor and Gradient Boosting is
very small.

Therefore:

**Stacking Regressor is the numerical winner by RMSE and R², while
Gradient Boosting is an extremely competitive alternative and has
the lowest MAE.**

The final report should not describe this difference as a
statistically significant improvement because no statistical
significance test was performed.
"""))


# ============================================================
# G14 — GENERALIZATION & OVERFITTING
# ============================================================

display(Markdown("""
## G14 — Generalization & Overfitting Analysis

Generalization analysis shows that training performance alone is
not sufficient for selecting a machine-learning model.

The Decision Tree achieved:

- Training Accuracy: **1.0000**
- Test Accuracy: **0.6115**

This produces a very large generalization gap and is strong evidence
of overfitting.

Bagging also achieved 1.0000 training accuracy but improved the
test Accuracy to 0.7087, demonstrating how ensemble aggregation
can improve generalization.

XGBoost had a substantially smaller gap:

- Training Accuracy: **0.8094**
- Test Accuracy: **0.7365**

AdaBoost showed the smallest observed classification accuracy gap:

- Training Accuracy: **0.7325**
- Test Accuracy: **0.7250**

These results demonstrate why unseen test performance and
cross-validation should be considered alongside training
performance.
"""))


# ============================================================
# G15 — CROSS-VALIDATION INTERPRETATION
# ============================================================

display(Markdown("""
## G15 — Cross-Validation Interpretation

Five-fold cross-validation was used to provide additional evidence
about model stability and generalization.

The cross-validation results broadly support the test-set findings.

For classification, the strongest models remained concentrated
among the boosting and ensemble approaches.

For regression:

- Gradient Boosting CV RMSE: **9.7561**
- Stacking Regressor CV RMSE: **9.7447**

The difference is only **0.0114** in cross-validation RMSE.

This reinforces the conclusion that Stacking and Gradient Boosting
have very similar regression performance.

Cross-validation therefore prevents the final conclusion from
depending entirely on a single train-test split.
"""))


# ============================================================
# G16 — FEATURE IMPORTANCE INTERPRETATION
# ============================================================

print("\n" + "=" * 90)
print("G16 — FEATURE IMPORTANCE INTERPRETATION")
print("=" * 90)

display(Markdown("""
## Important Predictive Factors

Feature-importance analysis from the boosting models identified
several variables that contributed strongly to prediction.

Important predictors included:

- Average quiz score
- Attendance rate
- Time spent on the course
- Number of sessions
- Videos watched
- Assignments submitted
- Quiz attempts
- Age
- Course timing features

The repeated appearance of assessment performance, attendance and
learning-engagement variables across models suggests that these
features contain useful predictive information.

### Important Caution

Feature importance indicates predictive contribution within the
trained model. It does **not** prove that a feature causes the
outcome.

For example, a high feature importance for attendance does not by
itself establish that increasing attendance will necessarily cause
a particular increase in final score.
"""))


# ============================================================
# G17 — BUSINESS / PRACTICAL INTERPRETATION
# ============================================================

display(Markdown("""
## G17 — Business & Practical Interpretation

The Smart Outcome Predictor can be interpreted as an early-warning
and academic-support system.

### Potential Applications

#### 1. Student Risk Identification

The classification model can help identify students who may be at
risk of not completing a course.

#### 2. Academic Intervention

Institutions could use prediction outputs to identify students who
may benefit from additional academic support.

#### 3. Performance Monitoring

The regression model can estimate expected final scores and help
instructors identify students whose predicted performance requires
attention.

#### 4. Engagement Monitoring

Variables such as sessions, time spent, videos watched, assignments
and attendance can provide useful signals for understanding student
engagement.

#### 5. Resource Allocation

Predictions could support targeted allocation of mentoring,
remedial learning material and instructor attention.

### Responsible Use

Predictions should support human decision-making rather than
automatically determine academic outcomes.

Model predictions should be interpreted together with instructor
knowledge and relevant student context.
"""))


# ============================================================
# G18 — COMPUTATIONAL EFFICIENCY
# ============================================================

display(Markdown("""
## G18 — Computational Efficiency

The experiments also considered training time.

The measured results showed:

- **XGBoost** was the fastest classification model among the
  boosting models tested.
- **LightGBM** was the fastest regression model among the boosting
  models tested.

These are observations from this experimental environment and
should not be interpreted as universal speed rankings.

Model selection should therefore balance predictive performance
with computational cost and deployment requirements.
"""))


# ============================================================
# G19 — HARD VOTING ROC-AUC NOTE
# ============================================================

display(Markdown("""
## G19 — Evaluation Methodology Note

For Hard Voting, the final class prediction is produced using
majority voting.

Because a Hard Voting classifier does not expose
`predict_proba()`, its ROC-AUC calculation uses the **mean
positive-class probability of the fitted base classifiers as a
continuous ensemble score**.

Therefore:

- Hard Voting class labels → majority voting
- Hard Voting ROC-AUC score → mean positive-class probability of
  the fitted base estimators

This distinction is explicitly documented to maintain evaluation
transparency.
"""))


# ============================================================
# G20 — LIMITATIONS
# ============================================================

display(Markdown("""
## G20 — Project Limitations

Although the project provides a comprehensive ensemble-learning
comparison, several limitations should be acknowledged.

### 1. Dataset Scope

The analysis is based on the available dataset and therefore may
not generalize to every educational institution or student
population.

### 2. Predictive Association

Feature importance represents predictive association rather than
causality.

### 3. Small Model Differences

Some top models have very similar performance. Small numerical
differences should not be interpreted as dramatic improvements.

### 4. No Statistical Significance Testing

The project does not perform formal statistical significance tests
between competing models.

### 5. External Validation

The models have not been evaluated on an independent external
dataset from another institution or learning platform.

### 6. Operational Deployment

The project evaluates predictive performance but does not implement
a production deployment pipeline or real-time monitoring system.
"""))


# ============================================================
# G21 — FINAL RECOMMENDATION
# ============================================================

display(Markdown("""
# 🎯 G21 — Final Recommendation

## Classification Recommendation

### 🥇 Recommended: Hard Voting

Hard Voting is selected as the preferred classification model for
this project because it achieved the strongest combination of:

- Accuracy
- F1-Score
- ROC-AUC

Its test-set results were:

**Accuracy = 0.7442**

**F1-Score = 0.6326**

**ROC-AUC = 0.7909**

XGBoost remains a strong alternative when Recall or simpler
single-model deployment is prioritized.

---

## Regression Recommendation

### 🥇 Numerical Winner: Stacking Regressor

Stacking Regressor achieved the best:

- RMSE = **9.8007**
- R² = **0.4859**

However, Gradient Boosting achieved the lowest:

- MAE = **7.8668**

Because the difference is extremely small, Gradient Boosting is a
practical alternative when simpler architecture and lower
complexity are preferred.

### Final Practical Choice

**Classification → Hard Voting**

**Regression → Stacking Regressor**

with **Gradient Boosting as the practical regression alternative**.
"""))


# ============================================================
# G22 — FINAL CONCLUSION
# ============================================================

display(Markdown("""
# 🏁 G22 — Final Conclusion

The Smart Outcome Predictor successfully demonstrated how ensemble
learning techniques can be applied to educational outcome
prediction.

The project progressed from data preparation and baseline models
through Bagging, Boosting, LightGBM, XGBoost, Voting and Stacking.

The experiments showed that ensemble methods can substantially
improve upon a single Decision Tree, particularly in terms of
generalization.

For classification, Hard Voting produced the strongest overall
held-out performance.

For regression, Stacking Regressor achieved the best RMSE and R²,
while Gradient Boosting achieved the best MAE.

The results also demonstrated that higher model complexity does not
automatically guarantee better performance. The small difference
between Stacking and Gradient Boosting highlights the importance of
considering computational cost and practical simplicity.

Overall, the project provides an evidence-based framework for
predicting course completion and final academic performance while
demonstrating the importance of appropriate evaluation metrics,
cross-validation, generalization analysis and responsible
interpretation.
"""))


# ============================================================
# G23 — COMPLETE PROJECT SCORECARD
# ============================================================

print("\n" + "=" * 90)
print("G23 — COMPLETE PROJECT SCORECARD")
print("=" * 90)

G_project_scorecard = pd.DataFrame({

    "Project Component": [

        "Data Understanding & Preparation",

        "Bagging",

        "Boosting",

        "LightGBM & XGBoost",

        "Voting & Stacking",

        "Evaluation Metrics",

        "Cross-Validation",

        "Generalization Analysis",

        "Visualization",

        "Final Interpretation"
    ],

    "Status": [

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed"
    ],

    "Evidence": [

        "Preprocessing pipeline + feature engineering",

        "Bagging Classifier + Regressor",

        "AdaBoost + Gradient Boosting",

        "Classification + Regression models",

        "Hard/Soft Voting + Stacking",

        "Accuracy, Precision, Recall, F1, AUC, MAE, RMSE, R²",

        "5-fold CV",

        "Train/Test gap analysis",

        "Confusion Matrix, ROC, Actual vs Predicted, Residuals",

        "Recommendations + limitations + conclusion"
    ]
})

display(
    G_project_scorecard
)


# ============================================================
# G24 — FINAL DELIVERABLES
# ============================================================

display(Markdown("""
# 📦 G24 — Final Project Deliverables

The completed project now contains the following major analytical
deliverables:

### Notebook

- Part A — Ensemble Learning Theory
- Master Library & Environment Setup
- Part B — Data Understanding & Preparation
- Part C — Bagging
- Part D — Boosting
- Part E — Voting & Stacking
- Part F — Evaluation Metrics & Model Comparison
- Part G — Final Report & Conclusions

### Analytical Outputs

- Cleaned dataset
- Classification comparison
- Regression comparison
- Cross-validation results
- Confusion matrices
- ROC curves
- Actual-vs-predicted visualization
- Residual analysis
- Feature-importance analysis
- Generalization analysis
- Final decision matrix
- Business interpretation
- Limitations
- Final recommendations
"""))


# ============================================================
# G25 — FINAL QUALITY GATE
# ============================================================

print("\n" + "=" * 90)
print("G25 — PART G FINAL QUALITY GATE")
print("=" * 90)

G_quality_checks = {

    "Executive summary present":
        True,

    "Project objectives documented":
        True,

    "Dataset summary documented":
        True,

    "Methodology documented":
        True,

    "Classification results available":
        len(G_classification_final) == 9,

    "Regression results available":
        len(G_regression_final) == 7,

    "Classification winners identified":
        (
            G_best_cls_f1["Model"] is not None
            and
            G_best_cls_auc["Model"] is not None
        ),

    "Regression winners identified":
        (
            G_best_reg_rmse["Model"] is not None
            and
            G_best_reg_r2["Model"] is not None
        ),

    "Limitations documented":
        True,

    "Business interpretation documented":
        True,

    "Final recommendations documented":
        True,

    "Final conclusion documented":
        True,

    "No missing classification metric values":
        G_classification_final[
            [
                "Accuracy",
                "Precision",
                "Recall",
                "F1-Score",
                "ROC-AUC"
            ]
        ].notna().all().all(),

    "No missing regression metric values":
        G_regression_final[
            [
                "MAE",
                "RMSE",
                "R²"
            ]
        ].notna().all().all(),

    "Project scorecard complete":
        len(G_project_scorecard) == 10
}


G_all_passed = True

for check_name, passed in G_quality_checks.items():

    print(
        f"{'✅ PASSED' if passed else '❌ FAILED'} | "
        f"{check_name}"
    )

    if not passed:
        G_all_passed = False


assert G_all_passed, (
    "Part G final quality gate failed."
)


# ============================================================
# G26 — FINAL PROJECT COMPLETION
# ============================================================

print("\n" + "=" * 90)
print("🏆 SMART OUTCOME PREDICTOR — PROJECT COMPLETED")
print("=" * 90)

print("""
╔══════════════════════════════════════════════════════════════════════════════╗
║                    🎉 FINAL PROJECT STATUS                                  ║
╠══════════════════════════════════════════════════════════════════════════════╣
║ Part A                         ✅ COMPLETE                                   ║
║ Master Library & Environment  ✅ COMPLETE                                   ║
║ Part B                         ✅ COMPLETE                                   ║
║ Part C — Bagging               ✅ COMPLETE                                   ║
║ Part D — Boosting              ✅ COMPLETE                                   ║
║ Part E — Voting & Stacking     ✅ COMPLETE                                   ║
║ Part F — Evaluation            ✅ COMPLETE                                   ║
║ Part G — Final Report          ✅ COMPLETE                                   ║
╠══════════════════════════════════════════════════════════════════════════════╣
║ Classification Recommendation: Hard Voting                                  ║
║ Regression Recommendation:     Stacking Regressor                            ║
║ Practical Regression Alternative: Gradient Boosting                         ║
╠══════════════════════════════════════════════════════════════════════════════╣
║                         🚀 READY FOR SUBMISSION                              ║
╚══════════════════════════════════════════════════════════════════════════════╝
""")
```
